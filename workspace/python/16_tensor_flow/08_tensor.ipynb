```python
model = nn.Sequential(
    nn.Linear(4, 6),
    nn.ReLU(),
    nn.Linear(6, 2)
)

print(model)

for m in model:
    print(f"\n --- {m} --- \n")
    try:
        print(m.weight.shape)
        print(m.bias.shape)
    except:
        print("no such attr")

print(f"\n --- model parameters --- \n")
for name, p in model.named_parameters():
    print(name, p)
```

```python
x = torch.randn(10, 4)

# 최종 형태는 반드시 (10, 2)
y = model(x)

# 4, 6
h1 = model[0](x)
# ReLU
h2 = model[1](h1)
# 6, 2
y2 = model[2](h2)

print(y)
print(y2)
```
