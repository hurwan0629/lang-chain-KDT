```python
import torch
print(torch.__version__)
print(torch.cuda.is_available())
```

```python
x = torch.tensor([
    [1.0],
    [2.0],
    [3.0],
    [4.0]
])

y = torch.tensor([
    [3.0],
    [5.0],
    [7.0],
    [9.0]
])
```

```python
model = torch.nn.Linear(1, 1)

criterion = torch.nn.MSELoss()

optimizer = torch.optim.SGD(
    model.parameters(),
    lr=0.01
)

print(list(model.parameters()))
```

```python
print(" ## 시작 ## ")
print("weight:", model.weight.item())
print("bias  :", model.bias.item())

for epoch in range(1, 201):
    # Linear() 객체 그 자체로 callable
    y_pred = model(x)

    # 손실계산
    # nn모듈의 MSELoss 클래스에 예측값과 실제값을 넣기만 하면 손실이 계산(집계)되어서 나옴
    loss = criterion(y_pred, y)

    # optimizer의 그래프 등록 객체들의 이전 기울기 초기화
    optimizer.zero_grad()

    # 각 변수들의 .grad에 대해서 편미분을 통해 기울기를 측정해주기
    loss.backward()

    # gradient를 이용해서 파라미터 업데이트
    optimizer.step()

    if epoch % 50 == 0:
        print(f"""


epoch : {epoch}
loss  : {loss.item()}
weight: {model.weight.item()}
bias  : {model.bias.item()}
""")

print(" ## 결과 ## ")
print("weight:", model.weight.item())
print("bias  :", model.bias.item())
```
