# zipfile 모듈
zipfile 모듈을 파이썬에서 제공하는 zipfile을 읽고, 만들고, 수정하고, 해제하기 위한 모듈입니다.

zipfile.ZipFile 클래스를 호출하여 인자로 zip파일 경로, 작업 이니셜 을 주게 됩니다.

> 이니셜의 종류로는 r(읽기), w(처음부터 쓰기), a(이어서 쓰기), x(새로 만들기) 가 있습니다.

해당 클래스는 위 과정을 통해 파일을 관리할 수 있는 객체를 반환해주게 됩니다.

```python
import zipfile

with zipfile.ZipFile("./data/shape.zip", "r") as zip_ref:
    # 위에서 설정한 경로로 zip 파일을 읽어 풀어주기
    zip_ref.extractall(extract_dir)
    print(list((extract_dir / "shape").iterdir()))
    print(list((extract_dir / "shape" / "train").iterdir()))
    print(list((extract_dir / "shape" / "test").iterdir()))
```

# ImageFolder
하위 폴더 이름을 클래스 이름으로 사용하며 알파벳 순서대로 클래스 번호를 지정하는 클래스입니다.

torchvisions.datasets.ImageFolder에서 제공해줍니다.

해당 클래스의 상속 관계는 다음과 같습니다.
1. Dataset
2. VisionDataset
3. DatasetFolder
4. ImageFolder

```python
from torchvision import datasets

train_dir = extract_dir / "shape" / "train"
test_dir = extract_dir / "shape" / "test"

raw_train = datasets.ImageFolder(train_dir)
raw_test = datasets.ImageFolder(test_dir)
```

```python
type(raw_train), type(raw_test)
```

```python
print(isinstance(raw_train, torch.utils.data.Dataset))
print(isinstance(raw_train, torchvision.datasets.VisionDataset))
print(isinstance(raw_train, torchvision.datasets.DatasetFolder))
print(isinstance(raw_train, torchvision.datasets.ImageFolder))
```

```python
# 데이터를 꺼내려하면 데이터를 즉시 로딩하여 반환
print(raw_train[0])
```

```python
# __getitem__ 에서 내부적으로 사용하는 samples에는 경로로 저장되어있음
raw_train.samples[0]
```

# 데이터 전처리와 증강
이미지 크기를 resize 하며 PIL Image를 수정 또한 지원하는 torchvision.transforms의 모듈을 이용하여 데이터를 수정하기에 올바른 형태로 수정할 수 있습니다.

또한 PIL 이미지는 모드 (L, RGB, RGBA, CMYK) 에 따라 채널이 다를 수 있기 때문에 이 또한 확인할 필요가 있습니다.

```python
from torchvision import transforms

train_transform = transforms.Compose([
    transforms.Resize((28, 28)),

    transforms.Grayscale(num_output_channels=1),
    transforms.RandomInvert(p=1.0),
    transforms.RandomAffine(degrees=12, translate=(0.08, 0.08), scale=0.90)
])
```
