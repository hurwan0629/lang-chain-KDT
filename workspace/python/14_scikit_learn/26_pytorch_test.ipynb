```python
torch.set_default_dtype(torch.float64)

x = torch.tensor([
    [1., 2.],
    [3., 4.]
])

with torch.set_grad_enabled(False):
    y = torch.nn.Linear(2, 1)(x)
    print(y)
    print(y.grad_fn)

with torch.set_grad_enabled(True):
    y = torch.nn.Linear(2, 1)(x)
    print(y)
    print(y.grad_fn)
    z = y ** 2
    print(z.grad_fn)
```
