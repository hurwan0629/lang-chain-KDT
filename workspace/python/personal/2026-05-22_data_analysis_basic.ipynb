```python
import pandas as pd

df = pd.DataFrame({
    "height": [170, 180, 160, 175, 168],
    "weight": [65, 80, 55, 72, 60],
    "label": ["normal", "heavy", "light", "normal", "normal"]
})

print(df)
print(df.describe())
print(f"키 평균: {df["height"].mean()}cm")
print(f"키 표준편차: {df["height"].std()}")
```

# 데이터 해석
- 데이터 개수: 5개
- 키: 170.6cm에 약 7.5의 표준편차를 가지고있다.
- 1,2,3사분위수: 168/170/175 
    - 약간 왼쪽으로 치우친 정규분포
- 무게는 대략 57~75 쯤에서 분포가 되어있다.

# 통계/수식 용어
- 평균: 전체 데이터의 중심 위치
    - `전체 합`/`데이터 개수`
- 편차: 특정 데이터가 평균에서 떨어진 정도
    - `특정 데이터 값` - `데이터 평균`
    - 편차를 모두 더하면 항상 `0`이 된다.
- 분산: 편차를 제곱해서 평균 낸 값
    - 데이터들이 얼마나 크게 분포되어있는지 확인하는 값
    - `[편차^2]`의 총합/데이터 개수
- 표준편차: 분산에 루트를 씌운 값
    - `(루트)분산`
    - np.std(ndarray)를 통해서 한번에 값을 뽑을 수 있다.

*학생들의 점수가 [50, 60, 75, 80, 95]인 경우*
- 평균: 총합(360)/5 = 72(점)
- 편차: [-22, -12, 3, 8, 23]
- 분산: (484 + 144 + 9 + 64 + 529)(=1230) / 5 = 246 -> 246
- 표준편차: 약 15점
> 이 점수들은 72점을 기준으로 15점정도 퍼져있다.

# 모집단 표준편차와 표본 표준편차
*표준 편차에는 보통 2가지가 존재한다.*
- 모집단: 전체 데이터를 다 가지고 있을 때 사용하는 방식
    - (설명과 동일)
- 표본 표준편차: 일부 샘플만 가지고 있을 때 사용 
    - m개 데이터중 n개만 존재한다면 `n개의 편차제곱합`/`n-1`
    - 일부 샘플만 보고 전체를 추정하지 않기 위하 분모에 `-1`을 보정해준다.


```python
import numpy as np

scores = np.array([60, 70, 85, 90, 100])
print(f"평균: {np.mean(scores)}")
print(f"표본분산: {np.var(scores, ddof=0)}")
print(f"표본표준편차: {np.std(scores, ddof=0)}")
print()
print("결과: 81을 기준으로 14만큼 분포되어있다. *표본기준")
```

```python
scores = np.array([60, 70, 85, 90, 100])

# 표본 방식
# ddof를 통해 설정 가능 (Delta Degrees of Freedom)
# numpy의 ddof기본값은 0 (모분산) *pandas는 ddof기본값이 1
print(f"평균: {np.mean(scores)}")
print(f"표본분산: {np.var(scores, ddof=1)}")
print(f"표본표준편차: {np.std(scores, ddof=1)}")
print()
print("결과: 81을 기준으로 14만큼 분포되어있다. *표본 기준")
```
