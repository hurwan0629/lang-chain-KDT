# 통계/수식 용어
- 평균: 전체 데이터의 중심 위치
    - `전체 합`/`데이터 개수`
- 편차: 특정 데이터가 평균에서 떨어진 정도
    - `특정 데이터 값` - `데이터 평균`
    - 편차를 모두 더하면 항상 `0`이 된다.
- 분산: 편차를 제곱해서 평균 낸 값
    - 데이터들이 얼마나 크게 분포되어있는지 확인하는 값
    - `[편차^2]`의 총합/데이터 개수
- 표준편차: 분산에 루트를 씌운 값
    - `(루트)분산`
    - np.std(ndarray)를 통해서 한번에 값을 뽑을 수 있다.

*학생들의 점수가 [50, 60, 75, 80, 95]인 경우*
- 평균: 총합(360)/5 = 72(점)
- 편차: [-22, -12, 3, 8, 23]
- 분산: (484 + 144 + 9 + 64 + 529)(=1230) / 5 = 246 -> 246
- 표준편차: 약 15점
> 이 점수들은 72점을 기준으로 15점정도 퍼져있다.

# 모집단 표준편차와 표본 표준편차
*표준 편차에는 보통 2가지가 존재한다.*
- 모집단: 전체 데이터를 다 가지고 있을 때 사용하는 방식
    - (설명과 동일)
- 표본 표준편차: 일부 샘플만 가지고 있을 때 사용 
    - m개 데이터중 n개만 존재한다면 `n개의 편차제곱합`/`n-1`
    - 일부 샘플만 보고 전체를 추정하지 않기 위하 분모에 `-1`을 보정해준다.


```python
import numpy as np

scores = np.array([60, 70, 85, 90, 100])
print(f"평균: {np.mean(scores)}")
print(f"표본분산: {np.var(scores, ddof=0)}")
print(f"표본표준편차: {np.std(scores, ddof=0)}")
print()
print("결과: 81을 기준으로 14만큼 분포되어있다. *표본기준")
```

```python
try:
  import numpy as np
except ImportError:
  pass


scores = np.array([60, 70, 85, 90, 100])

# 표본 방식
# ddof를 통해 설정 가능 (Delta Degrees of Freedom)
# numpy의 ddof기본값은 0 (모분산) *pandas는 ddof기본값이 1
print(f"평균: {np.mean(scores)}")
print(f"표본분산: {np.var(scores, ddof=1)}")
print(f"표본표준편차: {np.std(scores, ddof=1)}")
print()
print("결과: 81을 기준으로 14만큼 분포되어있다. *표본 기준")
```

# 사분위수 / IQR / 이상치
이전에 알아보았던 분산, 표준편차 등은 평균을 기준으로 퍼짐을 보는 방식이였습니다.

이번에는 데이터를 순서대로 세워서 위치 기준으로 퍼짐을 보는 **사분위수**와 **IQR**에 대해 알아보겠습니다.

---

# 사분위수
사분위수는 데이터를 작은값부터 큰 값까지 정렬한 뒤 4등분한 위치값을 말합니다.
이는 최대-최소 를 기준으로 나누는 것이 아닌 인덱스 순서대로 나누는 방식입니다.
인덱스가 `0`부터 시작된다고 한다면 (데이터는 `n`개입니다.) (소수점이 나온다면 내분점 공식을 이용합니다.)
- Q1의 인덱스: `(n-1)/4` 
- Q3의 인덱스: `(n-1)/4`*3  

가 됩니다.

그리고 해당 사분위수의 값은 `pandas`의 `Serires.quantile(0~1)`을 통해 확인 가능합니다. (0.25, 0.75) 

```python
try:
  import pandas as pd
except ImportError:
  pass

# len(series) = 8
series = pd.Series([10, 20, 30, 40, 50, 60, 70, 80])
# q1_index = (7/4) = 1.75 -> 20*(1/4) + 30*(3/4) = (20+90)/4 = 27.5
q1 = series.quantile(0.25)
# q3_index = (7/4)*3 = 5.25 -> 60*(3/4) + 70*(1/14) = 250/4 = 62.5
q3 = series.quantile(0.75)

print(f"Q1: {q1}, Q3: {q3}")
```

# IQR
IQR은 **Interquartile Range**의 줄임말로 **사분 범위**라고 부릅니다.
이는 단순하게 `Q3 - Q1`의 결과입니다.

# 이상치
이상치는 데이터중에서 너무 튀어서 일반적인 결과를 도출하는데 적합하지 않은 값입니다.

예를 들어서 `[1, 2, 3, 4, 900]` 이 있으면 여기에서 `900`을 이상치로 판단하기 합리적인 상황입니다.

`900`은 평균을 `2.5`에서 `182`로 올려버리게 됩니다.
중앙값은 이상치의 영향을 잘 받지 않아 `2.5`에서 `3`으로 큰 차이가 없습니다.

# IQR을 통한 이상치 판단하기 
사분 범위에서는 보통 아래와 같은 기준으로 이상치를 판단합니다.
- 하한선: Q1 - 1.5 * IQR
- 하한선: Q3 + 1.5 * IQR

이렇게 사분 범위보다 1.5배 벗어난 값은 이상치로 판단합니다.

코드에서 확인해보겠습니다.

```python
try:
  import pandas as pd
except ImportError:
  pass

print(f"\n --- init data --- \n")

data = pd.Series([10, 20, 30, 40, 50, 60, 70, 80, 900], dtype="float")

q1 = data.quantile(0.25)
q3 = data.quantile(0.75)
iqr = q3 - q1

lower_bound = q1- 1.5*iqr
upper_bound = q3+ 1.5*iqr
print(f"data: {data}")
print(f"mean: {data.mean()}")
print(f"median: {data.median()}")
print(f"q1: {q1}")
print(f"q3: {q3}")
print(f"IQR: {iqr}")
print(f"outlier range: lower than {lower_bound} or upper than {upper_bound}")

# outliers = map(lambda x: not (lower_bound <= x and x <= upper_bound), data.to_list())
# print(data[list(outliers)])

outliers = data[(data < lower_bound) | (upper_bound < data)]
print(f"outliers: {outliers}")

print(f"\n --- without delete outlier --- \n")

clean_data = data[(lower_bound <= data) & (data <= upper_bound)]
q1 = clean_data.quantile(0.25)
q3 = clean_data.quantile(0.75)
iqr = q3 - q1

lower_bound = q1- 1.5*iqr
upper_bound = q3+ 1.5*iqr
print(f"clean_data: {clean_data}")
print(f"mean: {clean_data.mean()}")
print(f"median: {clean_data.median()}")
print(f"q1: {q1}")
print(f"q3: {q3}")
print(f"IQR: {iqr}")
print(f"outlier range: lower than {lower_bound} or upper than {upper_bound}")
outliers = clean_data[(clean_data < lower_bound) | (upper_bound < clean_data)]
print(f"outliers: {outliers}")
```

# 평균/표준편차와 IQR의 비교
### 평균 기반 수치
평균/표준편차는 평균을 기준으로 얼마나 데이터들이 분산되어있는지를 확인하기 위해 사용하는 값입니다.
- **장점**
  - 수학적으로 다루기 좋고
  - 정규분포와 잘 연결되며
  - 머신러닝에서 많이 사용됩니다.
- **단점**
  - 이상치에 민감하다
### 사분위수/IQR
사분위수/IQR은 정렬된 위치를 기준으로 50%가 얼마나 퍼져있는지를 확인을 하며 이상치를 탐지하기 위해 쓰입니다.
- **장점**
  - 이상치에 강하다
  - 데이터의 실제 위치 감각을 보기 좋다
- **단점**
  - 평균/표준편차보다 수식 모델링에는 덜 적합합니다. (정렬 후 위치를 봐야하기 때문에)

`평균/표준편차/분산`과 `사분위수/IQR`은 모두 데이터의 상태를 확인하는 선행작업으로 이루어지기 위해 쓰이는 값들입니다.

`평균/표준편차/분산`은 민감하며 수학적으로 다루기 쉽고<br>
`사분위수/IQR`은 둔감하며 상대적으로 연산 비용이 더 들며 이상치를 탐지하기 좋습니다.
