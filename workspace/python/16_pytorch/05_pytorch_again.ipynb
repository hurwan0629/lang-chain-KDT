```python
import torch
```

```python
# 1. 학습 데이터
x = torch.tensor([1.0, 2.0, 3.0, 4.0])
y = torch.tensor([3.0, 5.0, 7.0, 9.0])

# 2. 학습할 파라미터
w = torch.tensor(0.0, requires_grad=True)
b = torch.tensor(0.0, requires_grad=True)

# 3. optimizer
optimizer = torch.optim.SGD([w, b], lr=0.01)

# 4. 학습
for epoch in range(100):

    # 예측
    # w와 함께 x에 대한 수식 -> [예측 값들 벡터로 변환]
    y_pred = w * x + b

    # MSE
    # 예측 데이터들이 얼마나 틀렸는지 확인하는 loss 지표 확인
    loss = ((y_pred - y) ** 2).mean()

    # 이전 gradient 제거
    # 최적화 알고리즘을 zero_grad를 이용하여 사용이 완료된 가중치들을 초기화 해주기
    optimizer.zero_grad()

    # gradient 계산
    # loss 수식을 미분하여 두 requires_gred가 있는 값에 대해 모두 편미분을 시킨 후
    # 현재 해당 값들을 대입해주게됨
    loss.backward()

    # 가중치 업데이트
    # 편미분 대상이 된 값들의 변화율에 학습률을 적용하여 파라미터들을 갱신해줌
    optimizer.step()

    # 10번에 한번씩 어떻게 값이 편하고있는지 찍어주기
    if epoch % 10 == 0:
        print(f"""
epoch={epoch:3d}
loss={loss.item():.4f}
w={w.item():.4f}
b={b.item():.4f}
""")

print("\n최종 결과")
print("w:", w.item())
print("b:", b.item())
```
