```python
import torch
```

```python
device = "cuda"

x = torch.tensor([1., 2., 3., 4.])
y = torch.tensor([3., 5., 7., 9.])
```

```python
w = torch.tensor(
    0.0,
    requires_grad=True,
    device=device
)#.to("cuda")
b = torch.tensor(
    0.0,
    requires_grad=True,
    device=device
)#.to("cuda")
```

```python
# 옵티마이저 생성
# torch의 optimizer 관련 모듈로 SGD, Adam과 같은 클래스를 제공해준다.
# SGD 옵티마이저 객체
# [w, b]는 파이썬 리스트로 학습해야할 파라미터를 주게됨
# lr은 learning_rate로 학습률을 의미함
optimizer = torch.optim.SGD([w, b], lr=0.01)
```

```python
print("target:", y)

for epoch in range(10000):

    # 예측
    y_pred = w * x + b

    # Loss 계산
    # x의 텐서는 4개짜리이기 때문에 이를 평균을 만들어주게 됨
    loss = ((y_pred - y) ** 2).mean()

    # 이전 gradient 제거
    # gradient가 누적합되지 않게 하기 위해 초기화 해주기
    optimizer.zero_grad()

    # gradient 계산
    # 여기에서 미분을 진행함
    loss.backward()

    # w, b 업데이트
    optimizer.step()

    if epoch % 1000 == 0:
        print(
            "\nepoch:", epoch,
            "\ny_target:", y,
            "\ny_pred:", y_pred,
            "\nloss item:", loss.item(),
            "\nw item:", w.item(),
            "\nb item:", b.item(),
            "\nloss:", loss,
            "\nw:", w,# tensor(1..., requires_grad=True) 같이 출력되지 않게하기
            "\nb:", b,
            "\n\n"
        )
```
