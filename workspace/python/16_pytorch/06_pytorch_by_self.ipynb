```python
import torch
```

```python
x = torch.tensor([1.0, 2.0, 3.0, 4.0])
y = torch.tensor([3.0, 5.0, 7.0, 9.0])

w = torch.tensor(0.0, requires_grad=True)
b = torch.tensor(0.0, requires_grad=True)
```

```python
optimizer = torch.optim.SGD([w, b], lr=0.01)
```

```python
for epoch in range(1, 201):

    # 먼저 존재하는 파라미터를 이용해서 예측해보기
    # x를 제외한 모든 데이터가 스칼라이기 때문에 그냥 곱하기 및 더하기로 연산 가능
    y_pred = w*x + b

    # 실제 정답과 비교하여 loss 점수 세우기
    # loss 점수는 벡터결과를 스칼라로 변환해야하기 때문에 .mean() 함수 사용해주기
    loss = ((y_pred - y) ** 2).mean()

    # 편미분 해주기 전에 모든 이전 기울기 결과를 초기화해주기
    optimizer.zero_grad()

    # loss 수식을 목적함수로써 미분 대상에 대해 모두 편미분해주기
    loss.backward()

    # 옵티마이저에 등록되어있는 모든 파라미터들을 기울기에 따라 학습시켜주기
    optimizer.step()

    # 50번에 단위로 출력해주기
    if epoch % 50 == 0:
        print(f"""
        epoch: [{epoch}/200]
target:{y}
y_pred:{y_pred}
w: {w.item()}
b: {b.item()}
""")

print("결과:")
print("w:", w.item())
print("b:", b.item())
```
