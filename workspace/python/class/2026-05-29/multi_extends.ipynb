# 파이썬의 다중상속

파이썬에서는 다중 상속을 지원하기 때문에 여러 메서드, 속성 등을 상속받아서 충돌이 일어날 수 있는데 이를 피하기 위해 **MRO**라는 선형 알고리즘을 이용하여 충돌을 회피한다.

**MRO**란 [부모] -> [자식1][자식2] -> [손자] 에서 자식1, 자식2의 충돌을 일어나지 않게 하는 것입니다.



```python
class Base:
    def hello(self):
        print("1. Base")
        
        print("Base끝")
        
class Clean(Base):
    def hello(self):
        print("1. Clean")
        super().hello()
        print("Clean끝")

class Pack(Base):
    def hello(self):
        print("1. Pack")
        super().hello()
        print("Pack끝")

class Product(Clean, Pack):
    def hello(self):
        print("1. Product")
        super().hello()
        print("Product End")

p = Product()
p.hello()

print(Product.mro())

# 상속받은 순서대로 
# Product(손주)의 helo -> 바로 위의 Clean()의 hello, 그 다음 순서인 Pack의 hello, 
# 마지막으로 super().hello()가 되어서 Base의 hello가 나온다.

# 이렇게 상속의 순서는
# Product > Clean > Pack > Base 가 된다.
```

# 다형성

다형성은 같은 이름의 메서드나 함수를 호출하더라도 실제 객체의 종류에 따라 다른 동작을 하게 하는 것을 의미합니다. 이는 일반적으로 부모의 기능을 오버라이딩 하는 클래스에서 나타나며 이것은 **동적 바인딩**을 통해 실협됩니다.

> 덕타이핑(duck typeing): 객체의 실제 클래스가 무엇인지 중요하지 않고, 필요한 메서드나 기능을 가지고 있으면 동작하는 방식

```python
from typing import Protocol

class Payable(Protocol):
    def pay(self, amount: int):
        ...

# 카드를 통한 결제
class CardPayment:
    def pay(self, amount):
        print(f"카드로 {amount}원 결제합니다")

# 현금을 통한 결제
class CashPayment:
    def pay(self, amount):
        print(f"현금으로 {amount}원 결제합니다")

# 카카오페이를 통한 결제
class KakaoPay:
    def pay(self, amount):
        print(f"카카오페이로 {amount}원 결제합니다")

class TossPay:
    def payment(self, amount):
        print(f"토스로 {amount}원 결제합니다.")

# 결제 방식을 통한 결제 (이때 payment는 .pay(amount) 메서드를 가지고 있는 클래스를 받는다.
def process_payment(payment: Payable, amount):
    payment.pay(amount)

card = CardPayment()
cash = CashPayment()
kakao = KakaoPay()
toss = TossPay()

process_payment(card, 10000)
process_payment(cash, 5000)
process_payment(kakao, 2000)
# process_payment(toss, 2000)
```

# 추상화
추상화는 복잡한 내부 구현은 숨기고, 외부에서는 필요한 핵심 기능만 간단하게 사용할 수 있도록 하는 개념입니다.

외부에서 사용하는 사용자는 객체가 내부적으로 어떻게 동작하는지 몰라도, 제공된 메서드 기능을통해 코드의 복잡도를 줄이고, 이해와 사용을 쉽게 할 수 있게 해줍니다.



```python
class ExampleClass():

  cls_var = "hello fellas" # public static String cls_var = "hello fellas";
  _cls_var = "better not use this" # 동일한 공개이지만 관례상 외부에서의 사용을 권장하지 않음
  __cls_var = "you can't see this" # 강하게 막지는 않지만 Name Mangling을 통해 직접 접근하기 어려운 방식이 됩니다.

  def __init__(self, name, password):
    self.name = name
    self.__password = password

  def _this_is_logic(self, message: str) -> str:
    return f"hello! I'm {self.name}. here is message: {message}"
  
  def hello_guys(self, message: str) -> str:
    return self._this_is_logic(message)

print(ExampleClass.cls_var)
print(ExampleClass._cls_var)
# print(ExampleClass.__cls_var)
print(ExampleClass._ExampleClass__cls_var)

e = ExampleClass("David", 1234)

print(e.name)
# print(e.__password) # 맹글링
print(e._ExampleClass__password) # 값이 나옴
print(e._this_is_logic("hello"))  # 사용 가능
print(e.hello_guys("good to meet you!"))


print(e.cls_var)
ExampleClass.cls_var = "changed"
print(e.cls_var)
e.cls_var = "again" # 인스턴스 클래스가 생성됨
print(e.cls_var)
print(ExampleClass.cls_var)
```
