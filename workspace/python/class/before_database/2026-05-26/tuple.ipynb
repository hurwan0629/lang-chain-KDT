# 3. 튜플
튜플은 여러가지의 데이터 타입의 값을 고정적으로 저장할 수 있는 순서가 있는, 변경 불가능한 순서가 있는 컬렉션입니다.
튜플은 보통 `,`을 이용하여 나타내면서 이를 `()`를 통해 감쌓게 됩니다.

예를 들어 `a = 1, 2, 3`으로 저장하게 된다면 `a`의 타입은 `tuple`이 됩니다.

```python
a = 1, 2, 3
print(type(a))
```

반대로 `(1)`과 같이 저장하게 된다면 해당 값은 `int(1)`이 됩니다.

```python
print((1) == int(1))
print((1, ) == int(1))
```

하지만 튜플에 `()`를 생략하며 선언이 가능하더라도 이를 `tuple`타입임을 나타내기 위해 `()`로 감싸주게 됩니다.

`a = 1, 2` 일 떄, `sum(a)`는 `sum((1, 2))`와 같습니다.

`print(sum(1, 2))`의 경우에는 튜플로 인식하지 않고, 인자를 2개 준 것으로 인식합니다.

```python
try:
  print(sum(1, 2))
except TypeError as e:
  print(e)
print(sum((1, 2)))
```

# 튜플의 저장 방식
튜플도 결국에는 불변 **객체**이기 때문에 데이터를 넣었을 때, 리스트와 같이 메모리 주소의 목록을 저장하게 됩니다.

예를 들어

```python
a = [1, 2, 3]
b = [4, 5, 6]
print(f"id(a): {id(a)}")
print(f"id(bs): {id(b)}")

t1 = a, b

print(f"(t1[0] is a): {t1[0] is a}")
```

위의 예시와 같이 `t1[0]`에 저장한 `a`는 두 값을 공유하게 됩니다.

또한 `a`의 값을 바꾸면 동일한 메모리를 참조하는 `t1[0]`또한 함께 바뀌게 됩니다. (사실 바뀐거라기 보다는 `t1[0]`이 보는 메모리의 객체가 바뀌었습니다.)

```python
a.append(999)

print(t1[0])
```

```python
import copy

original = list([[1,2], 2, 3])
print(f"id(original): {id(original)}")
print(f"original: {original}")
print(f"id(original[0]): {id(original[0])}")
print(f"original[0]: {original[0]}")

print(f"\n --- deep copy --- \n")

copy_deep = copy.deepcopy(original)
print(f"id(copy_deep): {id(copy_deep)} != {id(original)}(original) - 새로운 객체가 생긴다")
print(f"copy_deep: {copy_deep}")
print(f"id(copy_deep[0]): {id(copy_deep[0])} != {id(original[0])}(original) - 내부 요소의 메모리 주소가 다르다")
print(f"copy_deep[0]: {copy_deep[0]}")
print("\"실제 변수와 내부 데이터 모두 다른 변수로 대체된다.\"")

print(f"\n --- shallow copy --- \n")

copy_shallow = copy.copy(original)
print(f"id(copy_shallow): {id(copy_shallow)} != {id(original)}(original) - 새로운 객체가 생성된다.")
print(f"copy_shallow: {copy_shallow}")
print(f"id(copy_shallow[0]): {id(copy_shallow[0])} == {id(original[0])}(original)  - 내부의 메모리 주소는 같다.")
print(f"copy_shallow[0]: {copy_shallow[0]}")
print("\"새 객체를 만들되, 내부 객체는 재사용한다.\"")

```
