# 1. input 함수
input 함수는 사용자로부터 키보드 입력을 받는 데 사용되는 내장 함수입니다. 프로그램 실행 중에 사용자가 값을 입력하면, 입력한 값이 문자열로 반환되어 변수에 저장하거나 처리할 수 있습니다.

````python
import dis
code = """
s = input()

s
"""

compiled = compile(code, "<string>", "exec")
dis.dis(compiled)
print(compiled.co_code)
````

````python
try:
  import dis
except ImportError:
  pass



code = """
name, age, weight = input().split(' ')
print(f"{name} {age} {weight}")
"""

compiled = compile(code, "<string>", "exec")
dis.dis(compiled)
print(compiled.co_code)
````

# IO Stream에 대해서
`stream`은 데이터가 흐르는 통로를 말합니다.

파이썬은 직접 키보드나 화면을 만지는 것이 아닌 운영체제가 제공하는 입출력 통로(`stream`)을 통해 데이터를 주고 받습니다.

대표적으로 3가지의 `stream`이 존재합니다.
- `sys.stdin`: 표준 입력 (키보드 입력 통로) (input stream)
- `sys.stdout`: 표준 출력 (일반 출력 통로) (output stream)
- `sys.error`: 표준 에러 (에러 출력 통로) (output stream)

### `input()`는 `stdin`에서 읽게 됩니다.

`name = input("이름을 입력하세요: ")` 은 2가지 일을 합니다.
1. `"이름을 입력하세요: "`를 화면에 보여주기
2. 사용자가 입력한 값을 `stdin`에서 읽어오기

`input("string")`를 풀어서 쓰면 아래와 같습니다.

---
```python
import sys

sys.stdout.write("hello") # print("string", end="") # stdout
name = sys.stdin.readline() # stdin
name = name.rstrip("\n")
```

# 위의 코드 해설
터미널에서 파이썬을 실행하면 보통 하나의 구조가 만들어집니다. <br>
- Input Stream: 
  `[키보드] > [터미널 | 콘솔 | etc] > [운영체제 표준 io 통로] > [fd 0(stdin)] > [파이썬의 sys.stdin 객체] > [파이썬 문자열 str]` 
  <br><br>
- Output Stream: 
  `[파이썬 문자열 str] > [sys.stdout.write()] > [파이썬의 sys.stdoutt 객체] > [fd 1(stdout)] > [운영체제의 표준 출력 통로 > 터미널 화면]`





````python
import sys

print(sys.stdin)
print(sys.stdin.buffer)
print(sys.stdin.buffer.raw)
````
