```python
class ExampleBox():
  
  def __init__(self, prev, next, data):
    self.prev = prev
    self.curr = self
    self.next = next
    self.data = data

  def append(self, data):
    # 내가 맨 뒤면
    if self.next is None:
      self.next = ExampleBox(self, None, data)
    # 앞에 칸이 있으면 연결해주기
    else:
      # 기존 앞칸과 새 칸 연결해주기. (이때 내 prev는 아직 기존 앞칸과 연결되어있음)
      self.next.prev = ExampleBox(self, self.next, data)
      # 내 앞칸을 기존 앞칸의 next와 연결해주기
      self.next = self.next.prev
    
    return self.next
  
  def prepend(self, data):
    # 만약에 해당 박스가 맨 앞의 박스라면
    if self.prev is None:
      self.prev = ExampleBox(None, self, data)
    # 앞에 칸이 있으면 연결해주기
    else:
      # 기존 앞칸과 새 칸 연결해주기. (이때 내 prev는 아직 기존 앞칸과 연결되어있음)
      self.prev.next = ExampleBox(self.prev, self, data)
      # 내 앞칸을 기존 앞칸의 next와 연결해주기
      self.prev = self.prev.next
    
    return self.prev

  def __next__(self):
    if self.next is not None:
      return self.next
    raise StopIteration
      
  
class BoxSpace():

  def __init__(self, data):
    self.starting_point = ExampleBox(None, None, data)

  def __iter__(self):
    curr = self.starting_point
    
    while curr is not None:
      yield curr.data
      curr = curr.next
    # # # # [정확하지 않은 방식] # # # #
    # l = [curr.data]               #
                                
    # while (a := next(curr)):
    #   curr = a
    #   if a is not None:
    #     l.append(a.data)
    #   else:
    #     break

    # return iter(l)

b = BoxSpace(1)

import random

curr = b.starting_point

for i in range(10):
  r =  random.randint(i**2, i**4)
  curr = curr.append(r)

for d in iter(b):
  print(d)

print([i for i in iter(b)])



# print(next(next(b.starting_point)))

# s = iter([1, 2, 3, 4])

# while True:
#   if a := next(s):
#     print(a)
```
