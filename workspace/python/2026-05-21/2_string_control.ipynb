# String를 다루는 법
Python의 str은 문자들의 배열로 여러 메서드/함수가 존재한다.

```python
s1: str = "hello world?😍💚🎞️"


# 길이 반환
# 문자열 객체 안에 존재하는 length 메타데이터를 바로 읽음
len(s1)
```

```python
# 대문자 변환
# 각 문자 코드를 순회하면서 대문자 매핑이 있으면 바꿔 새 문자열을 만듦
print(s1.upper())
# 반대로
print(s1.lower())
```

```python
s1 = "   " + s1 + "   "
print(s1)
```

```python
# 양쪽 공백 제거
# 앞/뒤쪽에서 제거 대상 문자를 스캔 후 남은 범위로 새 문자열 생성
print(s1.strip())
print(str.strip(s1))
print(s1.lstrip())
print(s1.rstrip())
```

```python
# (문자열)arg1을 (문자열)arg2로 치환
# 원본을 왼쪽부터 탐색하며 a를 찾고 결과 버퍼에 b를 조립해 새 문자열 생성
print(s1.replace("😍", "❌❌❌"))
```

```python
# 위치 찾기
# 왼쪽부터 패턴 매칭 알고리즘으로 문자열이 시작하는 인덱스를 찾는다.
print(s1.find("😍💚"))
print(s1.rfind("😍💚"))
```

```python
# 위치 찾기
# find와 비슷함. 없으면 `valueError` 발생
try:
  print(s1.index("❌❌❌"))
except ValueError as e:
  print(e)
  
```

```python
# 첫 단어 대문자화 (나머지는 소문자)
# 문자/비문자 경계를 보면서 단어 시작 문자는 대문자, 나머지는 소문자로 만들기
print("hELLO worlD😍heaRt".title())
```

```python
# zfill(n)
# n길이에서 부족한 만큼 0을 추가하고 뒤에 기존 문자열을 추가한다.
str_num = "2"
print(str_num.zfill(2))
```

```python
# encode('utf-8')
# 내부 유니코드 코드 포인트를 지정한 인코딩 규칙에 따라 bytes 객체를 변환한다
print(s1.encode('utf-8'))
print(s1.encode('cp65001'))
```

```python
result = filter(lambda x: not(x[0:2]=="__" and x[-2:]=="__"), dir(s1))
print(type(result))
print(list(result))
```
