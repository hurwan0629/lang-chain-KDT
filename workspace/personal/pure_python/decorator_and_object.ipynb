```python
class Box:
  boxtype="박스입니다."

  def __init__(self, name):
    self.name = name

  def me(self):
    return self.name

  @classmethod
  def showClassType(cls):
    return cls.boxtype

  @staticmethod
  def sayHello():
    print("나는 박스야!")

Box.sayHello()

box1 = Box("종이박스")
box2 = Box("스티로폼 박스")

print(box1.showClassType())
print(box2.showClassType())

print(box1.me())
print(box2.me())

Box.boxtype="박스다"

print(box1.showClassType())
```
