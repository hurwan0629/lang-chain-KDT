# 파이썬의 함수와 `()`

파이썬의 함수는 일반적으로 `def [함수명](인자):` 형재로 만들어집니다. 이 때 `def func1():`으로 만들어진 함수는 `a=func1`, `a()`를 통해 실행이 가능합니다.

왜 `a=func1()`이라는 형태로 직관적인 대입을 하지 않을까요? (당연히 함수가 실행된 반환값을 대입받아서겠지만)

이는 함수 자체를 `a`에 대입하기 위해서입니다. 이는 함수의 활용 방법을 더 넓혀주게 됩니다.

한번 `func`를 찍어보겠습니다.

```python
def func1():
    return "hello"

a = func1()
print(a)

a = func1
print(a)

print(type.__class__)
print(int.__class__)
```

## 함수의 정체

이것도 사실상 `a = ExampleClass()`와 같다고 볼 수 있다.

이렇게 만들어진 객체는

# 함수 객체와 실행
`[함수 객체] + [()]`가 되면 해당 함수 객체 안의 `__code__`(바이트코드)를 실행한다라는 의미가 됩니다.

`()`가 붙으면 `CALL` 계열의 동작이 추가되게 됩니다.

`a()`를 실행 시 
1. `a`라는 이름 찾기
2. `a`가 가리키는 객체 가져오기
3. 객체가 호출 가능한지 확인
4. 함수 객체라면 안의 `__code__`를 실행할 준비 하기
5. 프레임 만들기
6. 매개변수/지역변수 공간 준비하기
7. `__code__`의 바이트코드 실행하기
8. `return`값 돌려주기

> `()`는 함수 뿐 아니라 **호출 가능한 객체** 모두에 대해서 실행을 하게 됩니다. 이 때문에 `def __call__(self)`가 존재하는 크르래스에서도 실행이 가능합니다.

```python
class A:
  def __call__(self):
    print("hello this is A")
  
  def __code__(self, a, b):
    a = 10
    b = 20
    print(a + b)
    return a + b
  
  def __str__(self):
    return "class A"
  
  
a = A()
a()
```

# 함수 호출 프레임

함수 객체는 계속 존재하고, `함수()`로 호출할 때마다 실행용 공간이 새로 생긴다.

이를 **스택 프레임** 또는 **함수 프레임** 이라고 부릅니다.

일반적으로
- 전역공간 (힙)
  - 함수 객체
  - 전역 변수
- 함수 프레임 (스택)
  - 함수의 지역 변수
  - 리턴값
  - (함수 끝나면 함수가 끝남)

이때, 함수 안에서 함수를 실행하게 된다면 프레임이 쌓이게 됩니다.

```python
def double(x):
    return x * 2

def add_and_double(a, b):
    total = a + b
    return double(total)

result = add_and_double(3, 4)
print(result)
```

위에서는 스택으로 
[전역 > add_and_double 프레임 > double 프레임]
