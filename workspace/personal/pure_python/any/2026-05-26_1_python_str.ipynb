# 1. 내가 아는 str
내가 아는 `Python`의 `str`은 문자셋, `C`언어의 `char`의 묶음으로 자동으로 관리가 되는 것이며, 내부적으로 불변객체로 존재하는 클래스 타입이다. (실제로는 유니코드 문자를 지정하며 단어 하나도 `char`보다 많은 것을 표현할 수 있다. 파이썬 내부적인 문자 저장 박식이 있다.)
이때분에 슬라이싱을 통한 부분 변환은 불가능하고, 메서드를 통해 새로운 객체를 생성하는 방식으로 사용되게 된다.

특별한 점은 `Python`은 메모리가 허용하는 한에서 길이의 제한이 없다는 것이다.

# 포메팅
포메팅은 변수에 값을 끼워넣는 방법으로 크게 3가지가 있다.
- `text = f"저의 이름은 {name} 입니다."`: `f`를 통해 `{}`에 변수를 넣는다. 자릿수 조절을 하고 싶은 경우에는 `{[변수]:2f}` 와 같은 형태로 작성이 가능합니다.
- `text = "저의 이름은 %s 입니다." % (name)`: 옛날 방식으로 `C`에서 주로 이런 방식을 사용한다. `%s`가 문자열, `%c`가 문자, `%.2f`가 실수, `%d`가 정수, `%r`이 개발자 표현(`repr()`) 등이다. `{값:포멧규칙}`으로 나타냅니다.
- `text = "저의 이름은 {} 입니다".format(name)`: `str.format()`을 통한 방식의 대입입니다. `{}` 안에 `0`, `1`, `...` 으로 순서 지정도 가능합니다.

```python
print(f"{12345.12345:20.4f}")
print(f"{1234:10d}")
print(f"가격: {242003:,}원")
print(f"{200:1.2%}")
print(f"{"hello":_^10}:<")
```
