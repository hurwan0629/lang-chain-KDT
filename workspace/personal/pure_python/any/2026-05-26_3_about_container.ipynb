# Iterable
이터러블은 반복 가능한 객체를 말합니다.
`for`문에 넣어서 하나씩 꺼낼 수 있는 객체입니다.

> `list`, `tuple`, `str`, `dict`, `set`, `range`, `generator` 등이 모두 `iterable`입니다. 시퀀스는 항상 `__iter__()` 메서드를 지원합니다.

이는 `__iter__()` 같은 메서드로 지원됩니다.


# Iterator
이터레이터는 실제로 `next()`로 값을 하나씩 꺼내는 객체입니다.
[파이썬 공식 문서](https://docs.python.org/3.10/library/stdtypes.html)에서 `Ctrl + F` 후 `StopIteration`를 찾아보면 이는 `__next__()`로 작동하며 마지막 값에서 `StopIteration` 예외를 발생시킨다고 나와 있습니다.

# Generator
값을 미리 다 만들지 않고 필요할 때 생성하는 `iterator`입니다.

동일하게 [파이썬 공식 문서](https://docs.python.org/3.10/library/stdtypes.html)에서 `Ctrl + F` 후 `Generator`를 써보면 설명에 자동으로 다음 값을 만들어 반환한다고 나와 있습니다.

예시로는 `(x * 2 for x in range(5))`와 같은 generator expression 등이 있으며 값은 `__next__()`가 호출될 때 지연 평가된다고 합니다.

> 이 때문에 제너레이터는 `match-case`의 시퀀스 패턴에서 일반 시퀀스처럼 매칭되지 않을 수 있습니다.

# Sequence
순서와 위치가 있는 `iterable`입니다.

`set`, `dict` 같은 자료구조도 `iterable` 객체이지만, `sequence`는 확실하게 인덱스와 순서가 존재하는 값입니다.


```
Iterable
├─ Sequence
│  ├─ list
│  ├─ tuple
│  ├─ range
│  └─ str
│
├─ Mapping
│  └─ dict
│
├─ Set
│  └─ set
│
└─ Iterator
   └─ Generator
```

# for문과 while문
`for`문과 `while`문은 모두 반복문이지만 (지금보니) 동작하는 구조가 꽤 다르다는 것을 알 수 있습니다.

## `while 조건:`
`while`문은 조건식이 `True`인 동안 반복하게 됩니다.

이때, 실행 구문 과정에서 조건에 대한 결과가 바뀌더라도 끝까지 실행 중인 구문을 완료한 뒤 조건문 체크에서 다시 체크를 하게 됩니다.

예를 들어 아래와 같은 코드는 10번 반복되게 됩니다.
```python
a = 0
i = 0
while a == 0 and i < 10:
  # True면 여기서부터 실행
  a += 1
  print(a)
  a -= 1
  i += 1
  # 여기까지 실행 후 다시 올라가기
# False면 여기로 이동
```

````python
a = 0
i = 0
while a == 0 and i < 10:
  a += 1
  print(a)
  a -= 1
  i += 1
````

# `for var in [iterable]`
`for`문은 내부적으로 `iterable`에서 값을 하나씩 꺼내는 반복문입니다.

흐름을 보면 아래와 같습니다.

````python
it = iter([10, 20, 30, 40, 50])

while True:
  try:
    x = next(it)
    print(x)
  except StopIteration:
    break
````

이 때문에 일반적으로 `for (int i = 0; i < 10; i++) {...}`와 같은 구조를 사용하는 전통적인 `for`문과 다르게 값을 하나씩 꺼내는 방식이기 때문에 `초깃값 설정 > 조건검사 > 증가 > 반복`과는 결이 미세하게 다릅니다.
