# 변수, 객체, 참조
파이썬은 객체 참조 방식을 사용합니다.

객체 참조 방식이란 값 그 자체에 데이터가 존재하는 방식이 아닌, 변수에 특정 **객체**의 메모리 주소를 저장한 뒤 해당 변수가 사용될 때 `LOAD`를 통해 내부적으로 연결된 객체의 데이터 값을 사용하는 방식입니다.

이 떄 파이썬은 변수를 내부적으로 프로세스 메모리의 `namespace`에 `{변수명: 가리키는 주소}` 형태로 저장하게 됩니다.

또한 아래에서 설명하듯 불변객체(`int`, `str`, `boolean` 등의 타입)들은 빠르게 접근 가능한 방식으로 저장하여 2개 이상의 불변 타입의 변수가 같은 값을 가진다면 둘은 같은 주소를 바라보게 만드는 방식을 사용합니다.

# `==`와 `is`
파이썬에는 두 변수가 동일한지 확인하는 방식은 크게 2가지 존재합니다.
- `==`: 클래스 내부에 설정된 `__eq__(self, other)`를 통해 비교를 합니다.
- `is`: 해당 클래스가 동일한 메모리를 바라보는지 확인하는 연산자입니다.

예를 들어서
```python
a = [1, 2, 3]
b = [1, 2, 3]

print(a == b)
print(a is b)
```
에서 내용물은 같기 때문에 `==`는 `True`, 서로 다른 객체이기 때문에 `is`는 `False`로 나오게 됩니다.

### 팁
- `a == b`는 내부적으로 `a.__eq__(b)`와 비슷하게 동작합니다. 리스트의 경우에는 첫번째 원소부터 순서대로 `==`을 통해 비교하는 방식을 활용합니다.
- `None`의 경우에는 파이썬에서 하나만 존재하는 **싱글톤 객체**이기 때문에 항상 모든 `None` 변수는 같은 메모리를 바라보기 때문에 오버라이딩 될 수 있는 `__eq__()`(`==`)보다는 `is`를 사용하는 것이 적절합니다.

````python
a = [1, 2, 3]
b = [1, 2, 3]

print(a == b)
print(a is b)
````

# 가변객체와 불변객체
앞에서 이미 한번 설명 설명하였지만 가변 객체를 한번 설명할겸 이야기를 해보겠습니다.
## 불변객체
불변객체는 이전에 설명하였기 때문에 간단히, `refcount`를 통해 자신을 참조하는 변수가 하나 이상 있을 경우에는 모든 같은 값을 가지는 불변 타입 변수들은 자신만을 사용하게 됩니다.

예를 들어서 `a, b = 10, 10`이라면 `(a is b) == True`가 됩니다.

이러한 객체 자체가 변하지 않는 특성 때문에 `a=10` 이후 `a=20`을 하면 `20 <class: 'int'>`객체가 생성되어 메모리에는 객체가 총 2개 (`a`가 보는 `20`과 `b`가 보는 `10`)가 존재하게 됩니다.

## 가변객체
반대로 가변 객체의 경우에는 바뀔 수 있는 객체를 말합니다.
이전 `a, b = 10, 10`에서는 `10` 하나를 `a`와 `b`가 바라보는 방식으로 실제 객체는 하나만 존재하였지만 <br>
`a, b = [1, 2, 3], [1, 2, 3]`의 경우에는 `list` 타입 객체가 2개가 생성되는 방식이 됩니다.

`a = [1, 2, 3]`후 `b = a`를 하면 두 변수는 하나의 객체를 바라보게 됩니다.

이때 `a.append(4)`를 하는 경우에는 새로운 리스트가 생성되는 방식이 아닌 `a`가 이미 바라보던 `[1, 2, 3] <class 'list'>`가 `[1, 2, 3, 4] <class 'list'>` 로 바뀌며 `namespace`의 `a`의 `value`인 메모리 주소는 바뀌지 않고 실제 객체의 형태가 바뀌게 됩니다.

### 사고해보기
- *튜플에 리스트가 들어가 있으면?* 튜플은 불변객체이지만 내부적으로 메모리 주소를 저장하기 때문에 내부의 리스트의 메모리 주소를 참조하여 값을 추가 및 삭제할 수 있다. 

## 추가 지식
같은 대상 바라보기: 위에서 말한 `b = a` 방식을 사용할 수 있습니다.
깊은 복사(Deep Copy): 리스트를 새로운 리스트에 다른 메모리의 같은 값으로 다른 객체를 만들기 위해서는 `copy` 모듈의 `deepcopy()` 함수를 사용해야합니다.
얕은 복사(Shallow Copy): 다른 리스트이지만 내부에서 바라보는 값은 같게 하기 위해서는 `b = copy.copy(a)`, `b = a[:]`이 도리 수 있습니다.

# 파이썬의 기본. 숫자 타입
파이썬에는 숫자타입이 몇가지 존재합니다.
- int: 일반 정수입니다. `C`나 `Java`와 다르게 자리수에 거의 제약이 없습니다. (메모리의 제약.)
- float: 실수입니다. 기본적으로 **IEEE 754***의 64비트 부동소수점 방식을 사용하여 미세한 값의 오차가 존재할 수 있습니다.
- bool: True/False 값입니다. 일반적으로 **데이터의 맥락상** 값이 `0`이면 `False`, 그 외에는 `True`로 나타냅니다. 실제로는 `bool`은 `int`의 하위 타입으로 동작하여 `True + True == 2`이며 `False + 10 == 10`, `isinstance(True, int) == True` 입니다.
- complex: 복소수입니다. 허수 단위는 `j`를 사용합니다. `1 + 2j` 같이 표현합니다. `.real`과 `.imag` 값이 존쟇바니다. 복소수 연산을 지원합니다.

이들은 실제로는 클래스와 동일한 방식으로 생성이 되며, 객체 참조 방식을 통해 변수에 저장이 됩니다.

Python의 실행 메모리 영역에는 변수가 가리키는 주소와 저장된 변수를 저장하는 `namespace`가 `{key: value}` 형태로 존재하며, 불변객체의 경우에는 id가 동일하게 저장이 되게 됩니다. (`int("1000")` 같이 직접 생성하는 경우 등에는 예외)

예를 들어서

```python
a = 10
b = 10
print(id(a) == id(b))
```

의 값은 True입니다.

````python

a = 10
b = 10
print(id(a) == id(b))
````

# str
`str`은 불변 객체인 문자들의 순서를 가진 **불변 시퀀스 객체**입니다.
이 때문에 인덱싱/슬라이싱은 가능하지만 리스트처럼 원소를 바꿀 순 없습니다.

````python
try:
  s = "hello world"
  print(s[0])
  s[0] = "H"
except TypeError as e:
  print("`s[0] = \"H\"`\n에서 오류가 났습니다")
````

# 연산자
이번 셀에서는 연산자 `+`, `-`, `*`, ... `or`, `not` 등등에 대해서 알아보겠습니다.

연산자의 경우에는 겉으로는 기호이지만 내부적으로는 타입별 연산 로직으로 연결되게 됩니다.

순서로 나열해보면 `연산자` > `바이트 코드` > `타입별 내부 연산 함수/매직 메서드로 연결` 되는 방식입니다.

> 이때 매직 메서드가 실행될 때 내장 타입은 내부 `C` 레벨 최적화 구현으로 처리되게 됩니다. 
> `is`, `and`, `or`, `not`는 매직메서드가 아닙니다.

````python
print((10).__add__(20))
print(10 and True)
print(True and 10)
print(10 or "hello")
print(not "hello")
````
