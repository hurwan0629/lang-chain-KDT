```python
import matplotlib.pyplot as plt

# 벡터 v = [3, 2]
x = 3
y = 2

plt.figure(figsize=(5, 5))

# 원점 (0, 0)에서 (3, 2) 방향으로 화살표 그리기
plt.quiver(
    0, 0,     # 시작점 x, y
    x, y,     # 벡터 방향 x, y
    angles='xy',
    scale_units='xy',
    scale=1
)

plt.xlim(-1, 5)
plt.ylim(-1, 5)

plt.axhline(0)
plt.axvline(0)
plt.grid(True)
plt.gca().set_aspect('equal')

plt.show()
```

```python
v = [3, 2, 4]

fig = plt.figure(figsize=(6, 6))
ax = fig.add_subplot(111, projection='3d')

# 원점에서 시작하는 3차원 벡터
ax.quiver(
    0, 0, 0,      # 시작점 x, y, z
    v[0], v[1], v[2],  # 방향 x, y, z
    length=1,
    normalize=False
)

ax.set_xlim(0, 5)
ax.set_ylim(0, 5)
ax.set_zlim(0, 5)

ax.set_xlabel('X')
ax.set_ylabel('Y')
ax.set_zlabel('Z')

plt.show()
```
