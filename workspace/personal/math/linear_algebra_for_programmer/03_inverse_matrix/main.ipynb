```python
# 높은행렬 (10, 4) 무작위로 만들기
T = np.random.randint(1, 100, size=(10, 4))
T_t = T.T

print(L := np.linalg.inv(T_t @ T)@T_t)

k = np.round(L@T, 2)
k = np.where(np.isclose(k, np.round(k)), np.round(k), k)
# k = np.trunc((np.linalg.inv(L)@T_t@T) * 100)/100
# print(k*100)
print(k)

print(np.round(T@L, 2))
```

```python
# 높은행렬 (10, 4) 무작위로 만들기
T = np.array([
  [1, 0],
  [0, 1],
  [1, 1]
])
T_t = T.T

print(L := T_t @ T)

np.linalg.inv(L)@T_t@T
```

```python
T = np.array([
  [1, 0, 3, 3+1],
  [0, 1, 4, 4+1],
  [1, 1, 8, 8+1]
])
try:
  print(np.linalg.inv(T))
except:
  print("np.linalg.inv(T) 실패")
finally:
  print(f"\n --- 의사 역행렬 --- \n")
  print(A_p := np.linalg.pinv(T))

np.round(T@A_p, 2)
```
