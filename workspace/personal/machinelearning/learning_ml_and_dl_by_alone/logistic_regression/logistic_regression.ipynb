```python
import pandas as pd

fish = pd.read_csv('https://raw.githubusercontent.com/rickiepark/hg-mldl/master/fish.csv')

fish.head()
```

```python
fish.info()
```

```python
X = fish[fish.drop(columns=['Species']).columns].to_numpy()
y = fish['Species'].to_numpy()
```

```python
fish.groupby(['Species'])['Species'].count()
```

```python
from sklearn.neighbors import KNeighborsClassifier
import numpy as np

kn = KNeighborsClassifier(n_neighbors=3)

kn.fit(X, y)

print(kn.classes_)

proba = kn.predict_proba(X[:5])
print(np.round(proba, decimals=4))
```

```python
from sklearn.preprocessing import StandardScaler

ss = StandardScaler()

bream_smelt_indexes = (y == "Bream") | (y == "Smelt")
train_bream_smelt = ss.fit_transform(X[bream_smelt_indexes])
target_bream_smelt = y[bream_smelt_indexes]
```

```python
from sklearn.linear_model import LogisticRegression

lr = LogisticRegression()
lr.fit(train_bream_smelt, target_bream_smelt)

print(lr.predict_proba(train_bream_smelt[:5]))
```

```python
print(lr.coef_, lr.intercept_)
```

```python
decisions = lr.decision_function(train_bream_smelt[:5])
print(decisions)
```
