# 선택 정렬과 삽입 정렬
선택 정렬과 삽입 정렬은 메커니즘이 닮아 있습니다.

선택 정렬은 뒤의 값들에서 최솟값을 하나씩 앞으로 보내는 방식이며 삽입 정렬은 앞에서부터 하나씩 올바른 위치에 놓는 방식입니다. 

예를 들어

`d a b e c`가 있으면 

선택 정렬의 경우에는
```
[a] [d b e c]
[a b] [d e c]
[a b c] [d e]
[a b c d] [e]
[a b c d e]
```
방식이고 삽입 정렬은
```
[d] [a b e c]
[a d] [b e c]
[a b d] [e c]
[a b d e] [c]
[a b c d e]
```
입니다.

결국 모두 랜덤값 기준 `O(n^2)`의 시간 복잡도를 가지지만 비운의 교육용 3대 정렬 알고리즘인 `버블`, `삽입`, `선택` 중에서 `삽입 정렬`은 이미 정렬되어 있는 경우 한 번의 순회만으로 추가 값을 넣을 수 있다는 점에서 실무에서 종종 쓰이기도 합니다.

> 역시나 잘 쓰이지 않는다는 점은 존재하지만 선택정렬 또한 **교환 횟수가 적다**라는 이점이 존재하여 교환 비용이 비싼 경우에는 유리합니다.

````python
# 삽입 정렬 구현
import random

# 20~30개 리스트 
# 1 ~ 2000 중 랜덤 고르기
rand_list = random.sample([a for a in range(1, 2001)], k=random.randint(20, 30))
print(rand_list)

order_type = "asc"
# order_type = "desc" # (desc가 아니어도 asc만 아니면 됨)

# 2번 만들기 귀찮아서.... 하핫
def insertion_sort(rand_list:list, order_type: str) -> list:
  # 횟수 세기용 count
  count = 0
  # 처음부터 하나씩 골라서 앞에 보내는 것이기 때문에 len(rand_list)-1까지 반복.
  # 하지만 0의 경우에는 이미 정렬되었다 가정. 따라서 1부터 시작
  for i in range(1, len(rand_list)):
    # print(f"{i}/{len(rand_list)-1}번째 순회")
    # print(rand_list)
    
    # 다음 값 뽑기
    pick = rand_list[i]
    j = i - 1
    # 기존 앞부분에서 알맞은 곳 탐색
    # 맨 뒤에서부터 자기의 알맞은 위치 찾아가기
    while j >= 0 and (rand_list[j] > pick if order_type == "asc" else rand_list[j] < pick):
      count+=1
      j -= 1
    
    # j+1 위치에 값을 넣으면 됨 (rand_list[j] < pick 인 상황이기 때문)
    # 나머지는 다 오른으로 밀어버리기
    rand_list[j+2:i+1], rand_list[j+1] = rand_list[j+1:i], pick

  print(rand_list)
  print(f"rand_list 길이: {len(rand_list)}")
  print(f"탐색 횟수: {count}")
  print(f"N^2 = {len(rand_list)**2}")
  return rand_list

print("\n --- asc --- \n")
insertion_sort(rand_list, "asc")
print("\n --- desc --- \n")
insertion_sort(rand_list, "desc")
````

````python
# 선택 정렬 구현
import random

# 20 ~ 30개 리스트 (중복이 많으면 더 줄 수 있음)
# 1 ~ 2000 중 랜덤 고르기
rand_list = random.sample([a for a in range(1, 2001)], k=random.randint(20, 30))
print(rand_list)

def selection_sort(rand_list:list, order_type: str) -> list:
  # 횟수 세기용 count
  count = 0

  # 무조건 순회 n*(n+1)/2
  for i in range(len(rand_list)):
    # print(f"{i+1}/{len(rand_list)-1}번째 순회")
    # print(rand_list)
    
    if order_type == "asc":
      # i번째 인덱스 값 구하기
      # 방법은 i부터 최솟값 탐색
      min = rand_list[i]
      min_index = i
      for j in range(i, len(rand_list)):
        count+=1
        if min > rand_list[j]:
          min = rand_list[j]
          min_index = j
      
      rand_list[i], rand_list[min_index] = min, rand_list[i]
    else:
      # i번째 인덱스 값 구하기
      # 방법은 i부터 최댓값 탐색
      max = rand_list[i]
      max_index = i
      for j in range(i, len(rand_list)):
        count+=1
        if max < rand_list[j]:
          max = rand_list[j]
          max_index = j
      
      rand_list[i], rand_list[max_index] = max, rand_list[i]

  print(rand_list)
  print(f"rand_list 길이: {len(rand_list)}")
  print(f"탐색 횟수: {count}")
  print(f"N^2 = {len(rand_list)**2}")
  return rand_list

print("\n --- asc --- \n")
selection_sort(rand_list, "asc")
print("\n --- desc --- \n")
selection_sort(rand_list, "desc")
    
````
