```python
# 메서드 = 주체가 생긴 함수
"hello".split("l")

# 객체.메서드()
class Coffee:
  def __init__(self, name, taste):
    self.name=name
    self.taste=taste
  
  def say_my_name(self):
    print(self.name)
    
  pass

# print(id(Coffee)) # Coffee(객체)의 주소를 반환함

c1 = Coffee # 클래스의 반환값은 자신의 주소
# print(id(Coffee))

c2 =  Coffee("c2", "맛") # ->
# c2.say_my_name()


# 메서드 시그니처
# 반환값 [자신의이름](인자)

c3 = Coffee("이름", "맛")
print(c3.say_my_name == c2.say_my_name)

id(c2)
```
