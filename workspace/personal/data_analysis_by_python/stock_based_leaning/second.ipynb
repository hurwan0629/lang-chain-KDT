```python
import pandas as pd
# Min-Max 정규화를 위한 자동화 함수
def normalize(df: pd.DataFrame):
  return (df - df.min()) / (df.max() - df.min())
```

```python
if False:
  NORMAL_NVDA = normalize(YF_NVDA.history(interval="1mo", start="2000-01-01"))
  NORMAL_SNP = normalize(STANDARD_AND_POOR.history(interval="1mo", start="2000-01-01"))
NORMAL_VIX = normalize(VIX.history(interval="1mo", start="2000-01-01"))
NORMAL_VIX
```

```python
import matplotlib.pyplot as plt

mo_index = NORMAL_SNP.index.tolist()

plt.figure(figsize=(12, 6))
plt.plot(mo_index, NORMAL_NVDA["Close"], label = "Nvdia")
plt.plot(mo_index, NORMAL_SNP["Close"], label = "S&P")
plt.plot(mo_index, NORMAL_VIX["Close"], label = "VIX")

plt.legend()
plt.tight_layout()

plt.show()
```
