```python
import pandas as pd
import numpy as np

orders = pd.DataFrame({
    "order_id": [
        "A001", "A002", "A003", "A004", "A005", "A006",
        "A007", "A008", "A009", "A010", "A011", "A012",
        "A013", "A014", "A015", "A016", "A017", "A018"
    ],
    "customer_id": [
        "C01", "C02", "C01", "C03", "C04", "C02",
        "C05", "C03", "C06", "C01", "C07", "C04",
        "C08", "C02", "C09", "C05", "C10", "C03"
    ],
    "order_date": [
        "2026-01-01", "2026-01-02", "2026-01-05", "2026-01-07",
        "2026-01-08", "2026-01-11", "2026-01-14", "2026-01-15",
        "2026-01-18", "2026-01-22", "2026-01-23", "2026-01-25",
        "2026-01-29", "2026-02-01", "2026-02-03", "2026-02-04",
        "2026-02-07", "2026-02-10"
    ],
    "product": [
        "notebook/pro", "mouse/basic", "keyboard/mechanical", "monitor/27inch",
        "notebook/air", "mouse/gaming", "chair/ergonomic", "monitor/32inch",
        "keyboard/basic", "notebook/pro", "desk/standing", "chair/basic",
        "monitor/27inch", "mouse/basic", "notebook/pro", "desk/normal",
        "keyboard/mechanical", "chair/ergonomic"
    ],
    "category": [
        "PC", "ACC", "ACC", "MONITOR", "PC", "ACC",
        "FURNITURE", "MONITOR", "ACC", "PC", "FURNITURE", "FURNITURE",
        "MONITOR", "ACC", "PC", "FURNITURE", "ACC", "FURNITURE"
    ],
    "quantity": [1, 2, 1, 1, 1, 3, 1, 1, 2, 1, 1, np.nan, 1, 2, 1, 1, 1, 1],
    "price": [
        1500000, 25000, 120000, 330000, 1100000, 55000,
        270000, 490000, 45000, 1520000, 380000, 90000,
        350000, 25000, 5000000, 220000, 130000, 280000
    ],
    "discount_rate": [
        0.05, 0.00, 0.10, 0.05, 0.03, 0.00,
        0.15, 0.07, np.nan, 0.05, 0.20, 0.00,
        0.05, 0.00, 0.50, 0.10, 0.10, 0.15
    ],
    "region": [
        "Seoul", "Busan", "Seoul", "Incheon", "Seoul", "Busan",
        "Daegu", "Incheon", "Seoul", "Seoul", "Busan", "Daegu",
        "Incheon", "Busan", "Seoul", "Daegu", "Seoul", "Incheon"
    ]
})

customers = pd.DataFrame({
    "customer_id": ["C01", "C02", "C03", "C04", "C05", "C06", "C07", "C08", "C11"],
    "grade": ["VIP", "Gold", "Silver", "Gold", "Silver", "Basic", "VIP", "Basic", "Gold"],
    "join_date": [
        "2025-03-01", "2025-07-15", "2025-09-10", "2025-11-20",
        "2025-12-01", "2026-01-01", "2025-01-10", "2026-01-20", "2025-08-08"
    ]
})
```

```python
orders["quantity"] = orders.groupby("category")["quantity"].transform("mean")
orders["quantity"]
```

```python
orders["discount_rate"] = orders["discount_rate"].fillna(0)
orders["discount_rate"]
```

```python


orders[["product_main", "product_sub"]] = orders["product"].str.split("/", expand=True).iloc[:, [0, -1]].set_axis(["product_main", "product_sub"], axis=1)
orders
```

```python
import datetime

print(orders[orders["order_date"] >= datetime.datetime(2026, 1, 15)])
```

```python
# 문제 2: 결측치 처리
# orders의 quantity 결측치를 category별 quantity 평균으로 채운 뒤, filled_orders 변수에 저장하세요.
# 힌트: groupby(), transform(), fillna()를 사용해보세요.

# 이미 되어있음.
orders["filled_orders"] = orders["quantity"]
orders
```

```python
# 문제 3: 할인 적용 금액 만들기
# orders에 final_price 컬럼을 추가하세요.
# 계산식: price * quantity * (1 - discount_rate)
# discount_rate 결측치는 0으로 처리한 뒤 계산하세요.

orders["final_price"] = orders["price"] * orders["quantity"] * (1 - orders["discount_rate"])
orders
```

```python
# 문제 4: 문자열 분리
# product 컬럼을 '/' 기준으로 나누어 product_main, product_sub 컬럼을 만드세요.
# 예: 'notebook/pro' -> product_main='notebook', product_sub='pro'

# 되어있음
```

```python
# 문제 5: 카테고리별 매출 집계
# category별 final_price 합계를 구하고, 매출이 큰 순서대로 정렬하세요.
# 결과는 category_sales 변수에 저장하세요.

category_sales = orders.groupby("category")["final_price"].sum()
category_sales
```

```python
# 문제 6: 지역별 주문 수와 평균 결제금액
# region별 주문 수(order_count)와 평균 final_price(avg_final_price)를 구하세요.
# 결과는 region_summary 변수에 저장하세요.

region_summary = orders.groupby("region")["final_price"].agg(["mean", "count"])
# agg({
#   "order_count": "sum",
#   "final_price": "mean"
# })
region_summary
```

```python
# 문제 7: 고객 정보 합치기
# orders와 customers를 customer_id 기준으로 left merge하세요.
# 결과는 order_customer 변수에 저장하세요.
# 주문에는 있지만 customers에 없는 고객도 사라지면 안 됩니다.

order_customer = pd.merge(orders, customers, how="left", on="customer_id")
order_customer
```

```python
# 문제 8: 등급별 매출 분석
# order_customer에서 grade별 final_price 합계와 평균을 구하세요.
# grade가 없는 고객은 'Unknown'으로 채운 뒤 집계하세요.
# 결과는 grade_sales 변수에 저장하세요.

grade_sales = order_customer.fillna({"grade":"Unknown"}).groupby("grade")["final_price"].agg(["sum", "mean"])
grade_sales
```

```python
# 문제 9: 월별 매출 만들기
# order_date에서 주문 월을 추출해 order_month 컬럼을 만드세요.
# 월별 final_price 합계를 구하고 month_sales 변수에 저장하세요.
# 힌트: dt.to_period('M') 또는 dt.strftime('%Y-%m')를 사용할 수 있습니다.

orders["order_month"] = orders["order_date"].dt.strftime("%Y-%m")
month_sales = orders.groupby("order_month")["final_price"].sum()
month_sales
```

```python
# 문제 10: 고액 주문 찾기
# final_price가 전체 final_price 평균보다 큰 주문만 필터링하세요.
# 결과는 high_value_orders 변수에 저장하세요.

print(orders[orders["final_price"] > orders["final_price"].mean()])
```

```python
# 문제 11: 고객별 구매 요약
# customer_id별 주문 횟수, 총 구매금액, 마지막 주문일을 구하세요.
# 컬럼명은 order_count, total_spent, last_order_date로 지정하세요.
# 결과는 customer_summary 변수에 저장하세요.

customer_summary = orders.groupby("customer_id")[["order_date", "final_price"]].agg(
  order_count=("final_price", "count"),
  total_spent=("final_price", "sum"),
  last_order_date=("order_date", "min")
)
```

```python
# 문제 12: 실전 조합 문제
# customer_summary와 customers를 합쳐 고객별 grade, join_date, order_count, total_spent를 확인하세요.
# 주문 이력이 없는 고객도 포함되게 만드세요.
# order_count와 total_spent의 결측치는 0으로 채우세요.
# 결과는 customer_report 변수에 저장하세요.

pd.merge(customer_summary, customers, on="customer_id", how="outer")[["grade", "join_date", "order_count", "total_spent"]].fillna({"order_count": 0, "total_spent": 0})
```
