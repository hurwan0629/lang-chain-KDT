```python
arr = np.array([5, 10, 15, 20, 25, 30, 100])

q1 = np.quantile(arr, 0.25)
q3 = np.quantile(arr, 0.75)

iqr = q3 - q1

print(iqr)

lower = q1 - 1.5* iqr
upper = q3 + 1.5* iqr

print(lower)
print(upper)
```

```python
# [fixed_acidity, volatile_acidity, alcohol, quality]
wine = np.array([
    [7.4, 0.70, 9.4, 5],
    [7.8, 0.88, 9.8, 5],
    [7.9, 0.76, 9.8, 6],
    [11.2, 0.28, 9.8, 6],
    [7.4, 0.70, 9.4, 5],
    [7.4, 0.66, 9.4, 5],
    [7.9, 0.60, 9.4, 5],
    [7.3, 0.65, 10.0, 7]
])

print(wine[wine[:, 3]>=6])
```

```python
print(wine[:, 2].mean())
```

```python
print(wine[wine[:, 3]>=6][:, 2].mean())
```

```python
good_wine_alcohol_mean = wine[wine[:, 3]>=6][:, 2].mean()

print(wine[wine[:, 2]>good_wine_alcohol_mean])
```

```python
print(wine[:, : -1])
```

```python
wine_feature = wine[:, :-1]

print(wine_feature.mean())
print(wine_feature.std())
```
