# Pandas와 메서드 활용
`pandas`의 최대 특징이라 한다면 아무래도 `DataFrame`가 아닐까 싶다.

`DataFrame`는 2차원 데이터 객체로 1개 이상의`Series`가 열 방향으로 붙여져있는 방식이다.

예를 들어 `s.name = "ticker"`이라는 이름의 시리즈가 있을 때, 순위로 인덱싱을 한다면
`s.index = ["first", "second", "third"]` 와 같이 설정하여 만들 수 있다.

예시는 아래와 같다.


```python
import pandas as pd

# 하나의 시리즈의 대표적인 설정 모습
st_ti = pd.Series(
  data=["QQQ", "NVDA", "GOOGL"], 
  index=["first", "second", "third"], 
  name="ticker"
)

print(st_ti)

print("\n --- add non str --- \n")

sample = st_ti.copy()
sample["fourth"] = 123
# dtype: object로 바뀜
print(sample)

print("\n bind two Series to DataFrame")

st_pr = pd.Series(
  data=[476.31, 5210, 4640], 
  index=["QQQ", "NVDA", "GOOGL"], 
  name="MarketCap(B)"
)

print("\n --- difference index (bad code) --- \n")

st_df = pd.concat([st_ti, st_pr], axis=1)
print(st_df)

print("\n --- same index (right code) --- \n")

st_pr.index = ["first", "second", "third"]

st_df = pd.concat([st_ti, st_pr], axis=1)
print(st_df)
```

# 복기
위와 같이, 시리즈는 배열 `name`과 `index`로 이루어진 테이블형 객체입니다.

이런 `Series`를 모아 저장하는 데이터 테이블이 `DataFrame`로 결합할 경우에는 `index`의 형태를 통일시켜주어야 합니다.

```python
print(st_df.dtypes)
```

# 값 찾기
`Series`와 `DataFrame` 모두 슬라이싱, 인덱싱 등이 가능하지만 더 많은 기능을 지원하기 때문에 일반 배열보다 사용법이 조금 더 복잡할 수 있습니다.

데이터를 더해보고 알아보겠습니다.

```python
import pandas as pd

data = {
    "ticker": [
        "QQQ", "NVDA", "GOOGL", "MSFT", "AAPL",
        "META", "AMZN", "TSLA", "AMD", "AVGO"
    ],
    "MarketCap(B)": [
        476.31, 5210.00, 4640.00, 3820.00, 3290.00,
        1580.00, 2360.00, 1120.00, 260.00, 1450.00
    ],
    "sector": [
        "ETF", "Semiconductor", "Communication", "Software", "Hardware",
        "Communication", "E-Commerce", "EV", "Semiconductor", "Semiconductor"
    ],
    "price": [
        456.20, 1042.00, 174.30, 421.50, 190.20,
        485.70, 184.80, 176.60, 161.40, 1410.30
    ]
}

index_names = [
    "first", "second", "third", "fourth", "fifth",
    "sixth", "seventh", "eighth", "ninth", "tenth"
]

st_df = pd.DataFrame(data, index=index_names)

print(st_df.info())
st_df
```

위와 같이 데이터(`DataFrame`)의 형태(`shape`)가 `(10, 4)`가 되었습니다.

이때 중간에 `third`~`eighth`의 `MarketCap(B)`, `sector`만 뽑는 방법을 알아보겠습니다.

```python
# .loc[[뽑을 인덱스(행)], [뽑을 속성(열)]]로 뽑기
print(st_df.loc[["third", "fourth", "fifth", "sixth", "seventh", "eighth"],
                ["MarketCap(B)", "sector"]])
# 일반 DF[]는 기본적으로 컬럼 선택만 가능합니다.
# 1. 컬럼 1개
print(st_df["ticker"][1:4])
```

```python
# 2. 컬럼 2개
print(st_df[["ticker", "price"]][5:])
```

```python
# 3. 특정 행만 출력
print(st_df[st_df["MarketCap(B)"] < 2000])
```

```python
# 행 슬라이싱
print(st_df[:5])
```

```python
# 행 라벨 슬라이싱
print(st_df["third":"ninth"])
```

```python
# DF 전체 슬라이싱 (일괄 불린마스크)

n_cols = ["price", "MarketCap(B)"]

n_mask = st_df[n_cols] > 200
s_mask_1 = st_df[["ticker"]].map(lambda x: 'A' in x) 
s_mask_2 = st_df[["sector"]] == "Semiconductor"

# print(n_mask)
# print(s_mask_1)
# print(s_mask_2)

print(st_df[pd.concat([n_mask, s_mask_1, s_mask_2], axis=1)])
```

```python
# loc - 이름 기준
print(st_df.loc[["third", "second"], ["sector", "price"]])

# iloc - 번호 기준
print(st_df.iloc[1:4, 2:4])
```

```python
# group by
print(st_df.groupby(["sector", "ticker"]).mean().sort_values("MarketCap(B)", ascending=False))
```
