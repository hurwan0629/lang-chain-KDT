# 판다스 연습문제 풀면서 익숙해지기

```python
import pandas as pd
import numpy as np


df = pd.DataFrame({
    "name": ["A", "B", "C", "D", "E", "F", "G"],
    "sector": ["AI", "AI", "Web", "Web", "Data", "Data", "AI"],
    "price": [10000, 15000, 8000, 12000, 20000, 18000, 13000],
    "sales": [5, 3, 10, 7, 2, 4, 6],
    "rating": [4.5, 4.0, 3.8, 4.2, 4.9, np.nan, 4.1]
})
```

```python
print(df.head(5))
```

```python
df.info()
```

```python
df.describe()
```

```python
print(df[["name","price", "rating"]])
```

```python
print(df[df["price"] > 12000])
```

```python
print(df[(df["sector"].eq("AI")) &  (df["price"] > 12000)])
```

```python
df["total"] = df["price"] * df["sales"]
df
```

```python
print(df.sort_values(by="total", ascending=False))
```

```python
print(df.groupby(by=["sector"])["price"].agg("mean"))
```

```python
print(df.groupby("sector")["sales"].mean())
```

```python
print(df.groupby(["sector"])[["price", "sales", "rating"]].agg({
  "price": "mean", 
  "sales": "sum", 
  "rating": "mean"
}))
```

```python
print(np.argwhere(df["rating"].isna()))
```

```python
df["rating"].fillna(df["rating"].mean())
```

```python
df["grade"] = np.where(df["rating"]>=4.2, "good", "normal")
```

```python
df
```

```python
print(df.loc[df["total"].idxmax()])
```

```python
print(df.groupby("sector")["total"].sum().sort_values(ascending=False))
```
