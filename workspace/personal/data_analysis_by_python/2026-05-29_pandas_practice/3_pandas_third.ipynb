```python
df["customer"] = df["customer"].str.strip().str.lower()
```

```python
df
```

```python
print(df[df["category"].isna()])
```

```python
df["category"] = df["category"].fillna("unknown")
df
```

```python
df["price"] = df["price"].fillna(df["price"].mean())
df
```

```python
df["quantity"] = df["quantity"].fillna(1)
```

```python
df["total"] = df["price"]*df["quantity"]
df
```

```python
df_valid = df[df["status"].eq("cancelled")]
print(df_valid)
```

```python
print(df[df["category"].str.contains("course", na=False)])
```

```python
df["category_simple"] = df["category"].str.replace("course", "").str.strip()
df
```

```python
print(df.groupby("customer")["total"].sum())
```

```python
print(df.groupby("category_simple")[["price", "quantity", "total"]].agg({
  "price": "mean",
  "quantity": "sum",
  "total": "sum"
}))
```
