```python
# dice
import random

dice_count = 1
dice_value_prob_int = {
  1: 10,
  2: 2,
  3: 2,
  4: 5,
  5: 8,
  6: 10
}

dice_list = []
dice_sum = 0
for i, v in dice_value_prob_int.items():
  dice_list += [i]*v
  dice_sum += v


def roll_dice():
  r = random.randint(0, dice_sum-1)
  return dice_list[r]
```

```python
result = {}
for i in range(1000):
  r = roll_dice()
  result[r] = result.get(r, 0) + 1

# print(result)
sum = 0
for _, v in result.items():
  sum += v

for i in range(1, 7):
  print(f"{i}: {int(result[i]/sum*100):2d}%")
```
