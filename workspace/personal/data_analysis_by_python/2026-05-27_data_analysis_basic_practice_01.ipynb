```python
import numpy as np
import pandas as pd

wine = pd.read_csv("../../resource/wine_resource/winequality-red.csv", sep=";")

print(wine.head())
```

```python
wine.info()
```

```python
desc = wine.describe()

condition = desc.loc["std"]/desc.loc["mean"] >= 0.4

big_std = desc[desc.columns[condition.tolist()]]

# print(big_std)

# 분포가 크다.
big_std_T = big_std.T
big_std_T["skew"] =  wine[big_std.columns].skew()
big_std_T["kurt"] =  wine[big_std.columns].kurt()

big_std_T

# print()

# big_std_T["skew"] = big_std_T[big_std.columns].skew()
# big_std_T["kurt"] = big_std_T.kurt()
# print(big_std_T)
```
