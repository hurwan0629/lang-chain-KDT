```python
df["prev_sales"] = df["sales"].shift(periods=1, axis=0)
df.head()
# df.info()
```

```python
df["is_above_avg"] = df["salary"] > df.groupby("department")["salary"].transform("mean")

df.head()
```

```python
pd.cut(
  df["score"], 
  bins=[0, 60, 70, 80, 90, 101], 
  labels=["F", "D", "C", "B", "A"], 
  right=False
)
```

```python
df.pivot_table(
  index = "region",
  columns = "category",
  values=["profit"],
  aggfunc="mean",
  fill_value=0
)
```

```python
df.sort_values(by="created_at").drop_duplicates(subset="email", keep="last")
```

```python
pd.concat([
  df[["store", "date", "sales"]],
  df.sort_values(["store", "date"], ascending=True).groupby("store")["sales"].cumsum().rename("cumsum_sales")
], axis=1).pivot_table(
  index="store",
  columns="date",
  values="cumsum_sales",
  aggfunc="first"
)
```

```python
pd.concat([
  df[["department", "salary"]],
  df.groupby(["department"])["salary"].rank(method="dense", ascending=False).rename("salary_rank")
], axis=1).sort_values(["department", "salary_rank"])
```

```python
import numpy as np

condlist = [
  ((df["sales"] >= 180) & (df["profit"] >= 30)),
  ((df["sales"] >= 120) & (df["profit"] >= 15)),
]

choices = ["high", "middle"]

pd.concat([
  df, 
  pd.Series(np.select(
    condlist,
    choices,
    default="low"
  ), name="performance")
], axis=1)
```
