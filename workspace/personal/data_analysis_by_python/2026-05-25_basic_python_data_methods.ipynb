```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt


df = pd.read_csv("../../resource/wine_resource/winequality-red.csv", sep=";")
target = df["quality"]
print(f"target: {type(target)}")
print(target)

print(df.describe())

```

```python
import pprint

result = dict()

result["mean"] = df.mean(numeric_only=True)
result["median"] = df.median(numeric_only=True) 
result["std"] = df.std(numeric_only=True) # = sqrt([df[numeric_colums] - df.mean(numeric_only)]**2/len(numeric_columns))
result["skew"] = df.skew() # 왜도
result["kert"] =  df.kurt() # 첨도
pprint.pprint(result)
print(f"\n --- \n")
pprint.pprint(result["mean"])
```

```python
s_residual_sugar = df["residual sugar"]

q1 = s_residual_sugar.quantile(0.25)
q3 = s_residual_sugar.quantile(0.75)
iqr = q3 - q1

upper = q3 + 1.5 * iqr
lower = q1 - 1.5 * iqr

residual_sugar_outlier_mask = (s_residual_sugar < lower) | (upper < s_residual_sugar)

print(s_residual_sugar[residual_sugar_outlier_mask])
```

```python
print(len(s_residual_sugar[residual_sugar_outlier_mask])/len(s_residual_sugar)*100)
```

```python
summary = {
    "outlier_count": [],
    "outlier_ratio": []
}

for col in df.select_dtypes(include="number").columns:
    s = df[col]
    q1 = s.quantile(0.25)
    q3 = s.quantile(0.75)
    iqr = q3 - q1

    lower = q1 - 1.5 * iqr
    upper = q3 + 1.5 * iqr

    mask = (s < lower) | (upper < s)

    summary["outlier_count"].append(mask.sum())
    ratio = mask.sum()/len(s)*100
    summary["outlier_ratio"].append(ratio)

    print(col + ": " + str(ratio)[0:4] + "%")
```

```python

df_outlier = pd.DataFrame(summary, df.columns)
print(df_outlier.head())
print(df_outlier.tail())

print(df_outlier[["outlier_ratio"]].sort_values("outlier_ratio", ascending=False))
```
