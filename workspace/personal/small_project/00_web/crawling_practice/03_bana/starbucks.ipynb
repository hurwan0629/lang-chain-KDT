```python
# import requests
from pathlib import Path
from selenium import webdriver
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC
from selenium.webdriver.chrome.options import Options
from selenium.webdriver.common.by import By
from selenium.webdriver.common.action_chains import ActionChains

import time
import re
import pandas as pd
from pathlib import Path
from urllib.parse import parse_qs, urlparse
```

```python
URL = "https://www.banapresso.com/"

options = Options()
options.add_argument("--start-maximized")

prefs = {
  "profile.default_content_setting_values.geolocation": 2,  # 위치
}

options.add_experimental_option("prefs", prefs)
options.add_argument("--disable-notifications")
```

```python
driver = webdriver.Chrome(options=options)

wait = WebDriverWait(driver, 5)

actions = ActionChains(driver)

driver.get(URL)

list_nav = wait.until(
  EC.presence_of_element_located(
    (By.CSS_SELECTOR, "li.list-item.depth")
  )
)

actions.move_to_element(list_nav).perform()

# 매장 찾기 화면으로 이동하기
wait.until(
  EC.element_to_be_clickable(
    (By.XPATH, '//a[span[normalize-space(text())="매장 찾기"]]')
  )
).click()


# 매장 리스트 목록 찾기
list_box_div = wait.until(
  EC.presence_of_element_located(
    (By.CSS_SELECTOR, "div.store_list_box > div.store_shop_list")
  )
)

# print(list_box_div.text)

# 팝업 끄기
wait.until(
  EC.element_to_be_clickable(
    (By.CSS_SELECTOR, "div.alert_popup button")
  )
).click()

# 스크롤 맨 아래로 내리기
while True:
  print("\n --- --- \n")
  prev_list_height = list_box_div.get_attribute("scrollHeight")
  print(prev_list_height)
  print(list_box_div.get_attribute("scrollTop"))
  driver.execute_script("arguments[0].scrollTop += arguments[0].scrollHeight;", list_box_div)
  time.sleep(0.3)
  print(list_box_div.get_attribute("scrollTop"))
  print(list_box_div.get_attribute("scrollHeight"))
  if prev_list_height == list_box_div.get_attribute("scrollHeight"):
    break
```
