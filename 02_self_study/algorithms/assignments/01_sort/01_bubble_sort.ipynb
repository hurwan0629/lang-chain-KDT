# 버블 정렬
버블 정렬은 일반적으로 실무에서 잘 쓰이지 않는 정렬 방식으로 교육을 위해 많이 쓰이는 방식입니다.

정렬의 이름에 존재하는 **버블**이라는 말과 같이 리스트의 앞부분부터 2개씩 순서대로 `[0, 1] [1, 2] [2, 3], ... [n-2, n-1]`을 정렬하는 것을 반복하여 최종적으로 정렬 상태를 만들게 됩니다.

최초의 실행에는 확실하게 `n-1`번째 인덱스가 제자리를 찾고, 그다음 실행에는 `n-2`가 확정되어 최종적으로 `n*(n-1) / 2`의 시간 복잡도, 즉 `O(n^2)`의 시간 복잡도를 가지며 공간 복잡도의 경우에는 그나마 교환을 위한 임시 저장 공간만을 요구하기 때문에 `O(1)`의 공간 복잡도를 소모합니다.

최선의 경우에는 `O(n)`으로 그냥 한 번 순회했을 때 정렬되어 있으면 그냥 패스하기도 합니다.

> 정렬에서 `O(n^2)`라는 시간 복잡도는 웬만하면 잘 나타나지 않는 최악에 가까운 시간 복잡도입니다.

```python
# 버블 정렬의 구현

import random

# 20 ~ 30개 리스트 (중복이 많으면 더 줄 수 있음)
# 1 ~ 2000 중 랜덤 고르기
rand_list = random.sample([a for a in range(1, 2001)], k=random.randint(20, 30))
print(rand_list)


def bubble_sort(rand_list:list, order_type: str):
  # 횟수 세기용 count
  count = 0

  # 버블 정렬
  # 마지막 한 번은 안 해도 문제없음
  for i in range(len(rand_list)-1):
    # 인덱스를 0 ~ (n-1)-i 까지 순회하며 정렬
    # 추가로 n-2, n-1 인덱스 교환만 하면 되기 때문에 n-2까지만 가면 됨
    
    # 중간에 정렬되었을 경우를 참고해서 한 번 검사해 주기
    swap = 0
    # print(f"{i+1}/{len(rand_list)-1}번째 순회")
    for j in range(len(rand_list) - i - 1):
      count += 1
      if rand_list[j] > rand_list[j+1] if order_type == "asc" else rand_list[j] < rand_list[j+1]:
          swap += 1
          # ▼ ▼ ▼ ▼ ▼ ▼ 파이썬의 좋은 점 ▼ ▼ ▼ ▼ ▼ ▼ 
          # 함수 하나를 안 만들어도 됨...
          rand_list[j], rand_list[j+1] = rand_list[j+1], rand_list[j]
    
    if swap == 0:
      print("정렬 완료")
      break

  print(rand_list)
  print(f"rand_list 길이: {len(rand_list)}")
  print(f"버블 횟수: {count}")
  print(f"N^2 = {len(rand_list)**2}")

print("\n --- asc --- \n")
bubble_sort(rand_list, "asc")
print("\n --- desc --- \n")
bubble_sort(rand_list, "desc")
```
