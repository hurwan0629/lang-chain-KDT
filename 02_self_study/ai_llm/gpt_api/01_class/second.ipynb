```python
from dotenv import load_dotenv
import os
import openai
from openai import OpenAI
import pprint

load_dotenv()
api_key = os.getenv("OPEN_AI_KEY")

print("OPEN_AI_KEY:", api_key[:8])

client = OpenAI(
    api_key=api_key,
)
```

```python
# messages = [
#     {
#         "role": "user", "content": "안녕하세요"
#     },
#     {
#         "role": "assistant",
#         "content": "안녕하세요! 무엇을 도와드릴까요"
#     },
#     {
#         "role": "user",
#         "content": "openai의 role 종류는 어떤것이 있고 각각 어떤 의미를 가지나요?"
#     }
# ]

messages = [
    {
        "role": "system", "content": "당신은 초등학교 선생님입니다. 한 답변당 토큰을 500개 이내로 쓰세요"
    },
    {
        "role": "user",
        "content": "openai의 role 종류는 어떤것이 있고 각각 어떤 의미를 가지나요? "
    }
]
```

```python
response = client.responses.create(
    model="gpt-5-nano",
    input=messages,
    max_output_tokens=2000,
    # temperature=0
)

print(response.status)
```
