```python
import gradio as gr
```

```python
with gr.Blocks() as demo:
    gr.Markdown("# 안녕하세요 Gradio!")
    gr.Markdown("### 작은 제목을 입력합니다.")
    gr.Markdown("- 첫번째 아이템 \n 두번째 아이템\n")

demo.launch()
```

```python
def handle_input(test):
    return test

with gr.Blocks() as demo:
    text_input = gr.Textbox(label="문자열 입력", lines=1)
    output_text = gr.Textbox(label="출력")
    text_input.submit(lambda text: text+"주가 입력", inputs=text_input, outputs=output_text)

demo.launch()

```

```python


with gr.Blocks() as demo:
    checkbox = gr.Checkbox(label="개인정보 사용에 동의하겠습니까?")
    output_checkbox=gr.Textbox(label="출력") # "동의했습니다" if prev else "동의하지 않았습니다."
    checkbox.change(lambda prev: prev , inputs=checkbox, outputs=output_checkbox)

demo.launch()
```

```python
with gr.Blocks() as demo:
    fruit_dropdown = gr.Dropdown(label="과일", choices=["사과", "오렌지", "키위", "바나나"])
    output_fruit = gr.Textbox(label="구입한 과일")
    fruit_dropdown.change(lambda fruit: f"선택한 과일: {fruit}", inputs=fruit_dropdown, outputs=output_fruit)

demo.launch()
```

```python
def add_func(num1, num2):
    return num1 + num2

interface = gr.Interface(
    fn=add_func,
    inputs = ["number", "number"],
    outputs = "number",
    title = "계산기",
    description = "숫자 두 개를 입력하세요",
    # flagging_mode = "never"
)
interface.launch()
```

```python
def favorite_language(language):
    messages = {
        "Python": "데이터 과학, 웹 개발, AI에 아주 적합한 언어",
        "JavaScript": "웹 개발에 강력하며, 프론트엔드와 백엔드에서 모두 사용 가능",
        "Java": "안정성과 성능으로 유명하며 대규모 시스템에 적합하다.",
        "C++": "고성능 어플리케이션과 게임 개발에 자주 사용"
    }

    return messages.get(language, "선택된 언어에 대한 정보가 없음")

interface = gr.Interface(
    fn =  favorite_language,
    inputs= gr.Radio([
        "Python",
        "JavaScript",
        "Java",
        "C++",
    ], label="좋아하는 언어"),
    outputs="text",
    title="좋아하는 언어",
    description="라디오 버튼에서 좋아하는 프로그래밍 언어를 선택하세요"
)

interface.launch()
```
