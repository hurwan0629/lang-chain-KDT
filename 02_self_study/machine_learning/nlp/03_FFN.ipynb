```python
import torch
import torch.nn as nn

x = torch.tensor([
    [
        [1.0, 2.0, 3.0, 4.0],
        [2.0, 1.0, 0.0, 1.0],
        [0.5, 1.5, 2.5, 3.5]
    ]
])

print(x.shape)
```

```python
linear1 = nn.Linear(
    in_features=4,
    out_features=8
)

hidden = linear1(x)

print(hidden.shape)
```

```python
[(n, p) for n, p in linear1.named_parameters()]
```

```python
activation = nn.ReLU()

hidden2 = activation(hidden)

print(hidden2.shape)
```

```python
print(hidden)
print(hidden2)
```

```python
linear2 =  nn.Linear(8, 4)

output = linear2(hidden2)

print(output.shape)
```

```python
class FFN(nn.Module):
    def __init__(self, dim_model, dim_ff):
        super().__init__()

        self.net = nn.Sequential(
            nn.Linear(dim_model, dim_ff),
            nn.GELU(),
            nn.Linear(dim_ff, dim_model)
        )

    def forward(self, x):
        return self.net(x)
```

```python
token = [x[:, i, :] for i in range(3)]

print(token[0].shape)
```

```python
ffn = FFN(4, 8)
o1 = ffn(token[0])
o2 = ffn(token[1])
o3 = ffn(token[2])

print(o1)
print(o2)
print(o3)
```
