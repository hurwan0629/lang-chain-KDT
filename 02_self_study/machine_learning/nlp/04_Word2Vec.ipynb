```python
from pathlib import Path
import os

NOVEL_FILE = Path().resolve() / "현진건-운수_좋은_날+B3356-개벽.txt"
```

```python
with open(NOVEL_FILE, "r", encoding="utf-8") as f:
    text = f.read()

print(text[:100])
```

```python
type(text), len(text)
```

```python
sentences = text.replace("\n", " ").split(".")

print(len(sentences))
print(sentences[:5])
```

```python
for s in sentences[:5]:
    print(s.strip())
    print("---")
```

```python
from mecab import MeCab

mecab = MeCab()
```

```python
sentence = sentences[2].strip()
print(mecab.pos(sentence))
```

```python
# 일반 명사 / 고유 명사 / 동사 / 형용사
keep_pos = ["NNG", "NNP", "VV", "VA"]
```

```python
tokens = []

for word, pos in mecab.pos(sentence):
    if pos in keep_pos:
        tokens.append(word)

print(tokens)
```

```python
tokenized_sentences = []

for sentence in sentences:
    sentence = sentence.strip()

    if not sentence:
        continue

    tokens = [
        word
        for word, pos in mecab.pos(sentence)
        if pos in keep_pos
    ]

    if tokens:
        tokenized_sentences.append(tokens)
```

```python
len(tokenized_sentences), sum([len(s) for s in tokenized_sentences]), tokenized_sentences[0]
```

```python
len(max(tokenized_sentences, key=lambda x: len(x)))
```

```python
from gensim.models import Word2Vec

cbow_model = Word2Vec(
    sentences=tokenized_sentences,
    vector_size=50, # 단어 하나를 50차원 벡터로 표현
    window=3, # 좌우 윈도우
    min_count=2, # 최소 문장 길이
    sg=0, # CBoW
    epochs=100
)
```

```python
# vocabulary 크기
len(cbow_model.wv)
```

```python
cbow_model.wv.index_to_key[:10]
```

```python
# token 김에 대한 벡터
cbow_model.wv["김"] == cbow_model.wv[1]
```
