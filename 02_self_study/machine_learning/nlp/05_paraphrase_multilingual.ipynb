# 사용해보기

```python
docs = [
    "자바에서 JVM은 자바 코드가 실행되기 위한 런타임 환경이다.",
    "json은 데이터를 구조적으로 저장하기 위한 형식중 하나이다.",
    "이진 분류 알고리즘인 퍼셉트론은 1957년 로젠블랫에 의해 구현되었습니다.",
    "파이썬의 리스트는 여러 개의 값을 순서대로 저장할 수 있는 자료구조이다.",
    "HTTP는 클라이언트와 서버가 데이터를 주고받기 위해 사용하는 통신 프로토콜이다.",
    "데이터베이스의 인덱스는 검색 속도를 향상시키기 위해 사용된다.",
    "트랜스포머는 어텐션 메커니즘을 기반으로 입력 데이터의 문맥을 처리한다.",
    "머신러닝에서 과적합은 학습 데이터에는 잘 맞지만 새로운 데이터에는 성능이 떨어지는 현상이다.",
    "Docker는 애플리케이션과 실행 환경을 컨테이너 단위로 패키징하는 기술이다.",
    "Git은 소스 코드의 변경 이력을 관리하기 위한 분산 버전 관리 시스템이다."
]

docs_vec = model.encode(docs, normalize_embeddings=True)
type(docs_vec), docs_vec.shape
```

```python
def search_doc(query):
    query_vec = model.encode(query, normalize_embeddings=True)

    similarities = (docs_vec @ query_vec)

    topk = np.argsort(-similarities)[:3]

    for idx, (doc, similarity) in enumerate([(docs[k], similarities[k]) for k in topk], start=1):
        print(f"[idx] [유사도 - {similarity:.4f}]: {doc}")
```

```python
query1 = "퍼셉트론은 어떻게 구현되었어?"

search_doc(query1)
```

```python
query2 = "코드 버전 관리 어떻게 해?"

search_doc(query2)
```
