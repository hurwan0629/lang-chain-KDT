# 강화학습에 관하여
강화학습은 `현재 어떤 행동을 해야 미래에 얻을 보상의 총합이 가장 커질까?` 에 대한 정답을 찾아가는 과정입니다.

강화학습에는 에이전트가 행동을 결정하며 행동에 대해 반응하는 세계가 존재하는 형태로 학습이 됩니다.

강화학습을 구성하는 핵심 요소로는 다음과 같은 요소가 존재합니다.
- Agent: 행동을 선택하는 주체
- Environment: 상호작용하는 환경
- State: 현재 상황을 나타내는 정보
- Action: 선택 가능한 행동
- Reward: 행동의 결가를 평가하는 신호
- Policy: 상태에 따라 행동을 선택하는 규칙
- Value(V, Q): 미래의 보상까지 고려한 가치
- Model: 환경의 변화와 보상을 예측하는 구조

에이전트는 한번의 상호작용에서 $s_t \xrightarrow[\text{Policy}]{a_t} r_{t+1}, s_{t+1}$ 와 같이 행동하게 됩니다.

이를 풀어서 설명해보면 
1. 현재 상태인 $s_t$가 존재합니다.
2. 정책 Policy에 따라 $a_t$ 행동을 선택합니다.
3. 해당 행동에 대한 보상 ($r_{t+1}$)과 다음 상태($s_{t+1}$)로 이어지게 됩니다.
