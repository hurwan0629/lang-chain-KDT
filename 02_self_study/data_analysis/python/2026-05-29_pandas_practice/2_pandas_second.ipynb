```python
import pandas as pd
import numpy as np

df = pd.DataFrame({
    "name": ["A", "B", "C", "D", "E", "F", "G", "H", "I", "J"],
    "sector": ["AI", "AI", "Web", "Web", "Data", "Data", "AI", "Web", "Data", "AI"],
    "level": ["junior", "senior", "junior", "middle", "senior", "junior", "middle", "senior", "middle", "junior"],
    "price": [10000, 15000, 8000, 12000, 20000, 18000, 13000, 9000, 22000, 11000],
    "sales": [5, 3, 10, 7, 2, 4, 6, 9, 1, 8],
    "rating": [4.5, 4.0, 3.8, 4.2, 4.9, np.nan, 4.1, 3.5, 4.8, np.nan]
})

df["total"] = df["price"] * df["sales"]
```
