```python
async def work(name, t):
    print("1" + name)
    await asyncio.sleep(t)
    print("2" + name)
    print(f"{name} 작업 끝")

def synchronousWork():
    print(4)
    time.sleep(1)
    print(8)
    print("b 작업 끝")

print(1)
await asyncio.gather(
    work("a", 4),
    work("b", 3),
    work("c", 1),
)
```
