# 변환행렬

```python
import numpy as np


x = np.array([1,2,3,4,5,6,7])
y = np.array([3,5,7,8,9,10, 13])

arr = np.array([
  np.concatenate((-1*x, x)),
  np.concatenate((-1*y, y))
])
```

```python
import matplotlib.pyplot as plt

T = [
  [1, 0.5],
  [0, 0.5]
]
k = T@arr

plt.scatter(arr[0], arr[1])
plt.scatter(k[0], k[1], color="red")
plt.show()


# plt.show()
```

```python


T = [
  [3, 0], # x축으로 *3
  [0, 0.5] # y축으로 반으로 나누기
]

k = T@arr

plt.scatter(arr[0], arr[1])
plt.scatter(k[0], k[1], color="red")
plt.show()
```

```python
gradient = 0.3

x = np.arange(-5, 30, 1)
y= x*0

df = np.array([x, y])

fig, ax = plt.subplots(figsize=(6, 6))

ax.scatter(x, y, color='blue', edgecolor='black', s=50, zorder=3, label='Data')

ax.set_xlim(-30, 30)
ax.set_ylim(-30, 30)

ax.axhline(0, color='black', linewidth=1.2)
ax.axvline(0, color='black', linewidth=1.2)

ax.grid(True, which='both', color='lightgray', linestyle='-', linewidth=0.5, zorder=1)

plt.show()
```

```python
radiant = np.pi/6

T = np.array([
  [np.cos(radiant), -np.sin(radiant)],
  [np.sin(radiant), np.cos(radiant)]
])

df_l = [T@df]
for i in range(1, 12):
  df_l.append(T @ df_l[-1])

# print(df_l)

fig, axes = plt.subplots(4, 3, figsize=(16, 12))


for i, ax in enumerate(axes.flat):

  ax.scatter(df_l[i][0], df_l[i][1], color='blue', edgecolor='black', s=50, zorder=3, label='Data')

  ax.set_title(f"{(i + 1) * 30}º rotate")

  ax.set_xlim(-30, 30)
  ax.set_ylim(-30, 30)

  ax.axhline(0, color='black', linewidth=1.2)
  ax.axvline(0, color='black', linewidth=1.2)

  ax.grid(True, which='both', color='lightgray', linestyle='-', linewidth=0.5, zorder=1)

plt.show()
```
