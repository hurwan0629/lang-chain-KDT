```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats
```

```python
# 베르누이 분포
p = 0.7
x = [0, 1]
y = stats.bernoulli.pmf(x, p)

plt.bar(x, y)
plt.xticks([0, 1])
plt.title("Bernoulli Distribution")
plt.xlabel("x")
plt.ylabel("Probability")
plt.show()
```

```python
fig, axes = plt.subplots(1, 3, figsize=(6, 2))

n_l = [10, 100, 1000]
p = 0.05

for ax, n in zip(axes, n_l):
  # 시행 횟수
  n = n
  # 성공 횟수
  p = 0.05

  x = np.arange(0, n+1)
  # 이항 분포
  y = stats.binom.pmf(x, n, p)

  ax.bar(x, y)

  mu = n * p
  # std = y.std()*3
  # print(y.mean())
  idx = np.where(y==y.max())[0][0]
  print(idx)
  # print(y.std()*3)
  # print(std)
  ax.set_xlim(max(0, mu-idx)-10, max(mu+idx, 10)+10)

  ax.set_title(f"n={n} / p={p}")
# fig.title("Binomial Distribution")

# fig.xlabel("Number of Successes")

# fig.ylabel("Probability")
fig.tight_layout()
fig.show()
```

```python
fig, axes = plt.subplots(1, 3, figsize=(9, 3))

lam_l = [2, 50, 100] # 단위 시간당 평균 발생 횟수

for ax, lam in zip(axes, lam_l):

  # 단위 시간당 x[i]번 오류가 일어날 배열
  x = np.arange(0, lam*3)
  # 평균이 lam일 때 단위 시간동안 일어날 확률
  # 포아송 분포
  y = stats.poisson.pmf(x, lam)

  ax.bar(x, y)
  ax.set_xlim(-10, max(lam, 10)*2)
  ax.set_title(f"{lam} events for unit time")
  ax.set_xlabel("Number of Events")
  ax.set_ylabel("Probability")


fig.tight_layout()
fig.show()
```

```python
# -1에서 시작해서 2 까지 500등분 (499분위 하기)
x = np.linspace(-1, 2, 500)
x = np.random.random(100) * 2
x.sort()
# print(x)
# 유니폼 분포
y = stats.uniform.pdf(x, loc=0.5, scale=1)

plt.plot(x, y)
plt.xlim(-1, 5)
plt.title("Uniform Distribution")
plt.xlabel("x")
plt.ylabel("Density")
plt.show()
```
