# 1. 사이킷 런
사이킷런은 파이썬으로 작성된 오픈소스 머신러닝 라이브러리로 데이터 분석과 예측 모델을 구축하기 위해 쓰입니다. 지도학습, 비지도학습 등을 모두 제공하며 모델 선택부터 차원축소, 클러스터링, 회귀, 전처리 등과 같은 데이터 과학 및 모델링에 쓰기 위한 많은 작업을 수행해주는 라이브러리입니다.

# 데이터셋
데이터셋은 머신러닝에서 데이터 과학에서 모델을 학습, 검증, 테스트를 하기 위해서 사용하는 데이터의 총합입니다.

보통 데이터는 학습 -> 검증 -> 테스트로 이루어지며 [학습/검증] 데이터는 보통 부트스트랩핑과 같은 방식을 통해 나누질 수 있습니다. 또한 모델이 만들어지는 도중에 학습이 될 하이퍼파라미터 조정이 있으며 품질과 크기에 큰 영향을 미치는 전처리, 특성 엔지니어링 등을 할 수 있습니다.

여기에서 중요한 것은 **학습량 데이터셋**인데 검증과 테스트 데이터셋을 제외한 데이터셋을 말합니다.

```python
from sklearn.datasets import load_iris
```

```python
iris = load_iris()
# iris

# sepal length in cm: 꽃받침의 길이
# sepal width in cm: 꽃받침의 너비
# petal length in cm: 꽃잎의 길이
# petal width in cm: 꽃잎의 너비
print(iris['DESCR'])
data = iris['data']

feature_names = iris['feature_names']


import pandas as pd

df_iris = pd.DataFrame(data, columns=feature_names)


target = iris['target']
target
```

```python
print(iris['target'])
```

```python
feature_names = iris['feature_names']
feature_names
```

```python
data
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    data, target,
    train_size=0.7,
    test_size=0.3,
    random_state=42,
)

```

# SVC

SVC(Support Vector Classifier)는 서포트 벡터 머신(SVM, Support Vector Machine)을 사용한 분류(Classification) 알고리즘입니다. 사이킷런(sklearn.svm.SVC)에서 제공되며, 이진 분류(Binary Classification)와 다중 클래스 분류(Multi-Class Classification) 문제를 해결할 수 있습니다.



서포트 벡터 머신(Support Vector Machine, SVM)

서포트 벡터 머신(SVM)은 두 개 이상의 클래스(Class)를 구분하는 지도 학습(Supervised Learning) 알고리즘입니다. 주로 분류(Classification) 문제를 해결하는 데 사용되며, 일부 경우 회귀(Regression) 문제에도 사용됩니다. SVM의 목표는 두 클래스 간의 경계를 가장 잘 구분하는 최적의 초평면(Hyperplane)을 찾는 것입니다.

```python
from sklearn.svm import SVC

svc = SVC()
svc.fit(X_train, y_train); # ; vscode에서 표현하지 못하는 이미지 생략
```

```python
svc.predict(X_test)
```



Accuracy

정확도(Accuracy)는 머신러닝과 통계학에서 모델의 성능을 평가하는 가장 기본적인 지표 중 하나로, 전체 예측 중에서 얼마나 많은 예측이 실제 정답과 일치했는지를 나타내는 비율입니다.
