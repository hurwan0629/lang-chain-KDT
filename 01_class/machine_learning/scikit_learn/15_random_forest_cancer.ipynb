# 유방암 분류 데이터셋 연습

```python
from pandas.conftest import ascending
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, confusion_matrix
from sklearn.model_selection import train_test_split
from sklearn.datasets import load_breast_cancer
```

```python
data = load_breast_cancer()

X = data.data
y = data.target

print("X.shape:", X.shape)
print("y.shape:", y.shape)

print(data.feature_names)
print(data.target_names)
```

# 데이터 간단한 확인
평균 반지름, 텍스쳐, 너비, 에러, 최악 부위 등이 존재하며 결과 특성 이름인 `malignant`는 악성을 의미하고 `benign`은 암이 아닌 양성을 의미합니다.

```python
X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.2, # 검증 데이터 20%
    random_state=42,
    stratify=y # 층화 추출
)
```

```python
model = RandomForestClassifier(
    n_estimators=100,
    random_state=42
)
```

```python
# 학습 및 예측
model.fit(X_train, y_train)
y_pred = model.predict(X_test)

print(type(y_pred), y_pred.shape, y_pred[:5])
```

```python
print(model.predict_proba(X_test)[:10])
```

```python
# 평가
print("accuracy:", accuracy_score(y_test, y_pred))

print("\nconfusion matrix:")
print([f"{name}: {score}" for name, score in zip(["TN", "FP", "FN", "TP"], confusion_matrix(y_test, y_pred).reshape((4,)))])
```

```python
# 모델 내부 확인
print("\nclasses:")
print(model.classes_)

print("\nfeature importances:")
print(model.feature_importances_)

print("\ntree_count:")
print(len(model.estimators_))
```

```python
important_features = []

for name, importance in zip(
    data.feature_names,
    model.feature_importances_
):
    # 평균보다 중요도가 높은 특성들만 추출
    if importance > model.feature_importances_.mean():
        important_features.append((str(name), str(importance)))

important_features.sort(key=lambda x: x[1])
important_features
```
