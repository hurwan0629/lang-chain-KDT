### SPRi PDF 기반 정책 산업 브리핑 어시스턴트
AI 전략팀에서 2026년 10월 SPRi AI Brief를 바탕으로 회의 준비를 돕는 챗봇을 만듭니다. 팀원들은 PDF를 직접 읽지 않고도 관련 페이지를 찾아 요약하고 싶어 합니다.

1. 첨부된 PDF만 검색 대상으로 사용하세요. 외부 웹 검색은 사용하지 않습니다.
2. Retriever를 사용해 질문마다 관련 청크 최대 4개를 가져오세요.
3. 답변에는 PDF 페이지 번호를 포함하세요.
4. API 키가 없을 때에도 검색 결과 페이지는 확인할 수 있어야 합니다.
5. 답변 근거가 없으면 추측하지 말고 확인 불가라고 답하세요.
6. 두 개 이상의 질문을 자동 테스트하고, 결과를 비교하세요.
7. BM25와 벡터 Retriever를 교체해 결과가 어떻게 다른지 확인하세요.

```python
from pathlib import Path
from typing import TypedDict
from langchain_community.document_loaders import PyPDFLoader
from langchain_text_splitters import RecursiveCharacterTextSplitter

pdf_path = next(Path(".").glob("SPRI*.pdf"))

pages = PyPDFLoader(str(pdf_path)).load()

splitter = RecursiveCharacterTextSplitter(
  chunk_size=600,
  chunk_overlap=100
)

chunks = splitter.split_documents(pages)
```

```python
for i, doc in enumerate(chunks):
  doc.metadata["doc_id"] = f"SPRi-{i:04d}"
  doc.metadata["page_display"] = doc.metadata["page"] + 1

print("페이지:", len(pages))
print("청크:", len(chunks))
```

```python
import re
from langchain_community.retrievers import BM25Retriever

def tokenize(text):
  return re.findall(r"[가-힣A-Za-z0-9]+", text.lower())

bm25 = BM25Retriever.from_documents(
  chunks,
  preprocess_func=tokenize
)

bm25.k = 4

docs = bm25.invoke("파이프라인")

for doc in docs:
  print(doc.metadata["page_display"],"쪽 |", doc.page_content[:100].replace("\n", ""))
```

```python
import os
from dotenv import load_dotenv
from langchain_openai import OpenAIEmbeddings
from langchain_chroma import Chroma

load_dotenv()

vector_retriever = None

if os.getenv("OPENAI_API_KEY"):
  embeddings = OpenAIEmbeddings(
    model="text-embedding-3-small"
  )

  db=Chroma(
    collection_name="spri",
    embedding_function=embeddings,
    persist_directory="./chroma_db"
  )

  db.add_documents(
    chunks,
    ids=[doc.metadata["doc_id"] for doc in chunks]
  )

  vector_retriever = db.as_retriever(
      search_kwargs={"k": 4}
  )
```

```python
from typing import TypedDict
from langchain_core.documents import Document
from langgraph.graph import StateGraph, START, END
from langchain_openai import ChatOpenAI

class State(TypedDict):
    query: str
    mode: str
    docs: list[Document]
    answer: str

# 검색 노드
def retrieve(state: State):
    retriever = bm25

    if state["mode"] == "vector":
        if vector_retriever is None:
            raise ValueError("벡터 검색용 API 키가 없습니다.")
        retriever = vector_retriever

    docs = retriever.invoke(state["query"])

    return {"docs": docs}


# 답변 노드
def answer(state: State):
    context = "\n\n".join(
        f"[PDF {doc.metadata['page_display']}쪽]\n{doc.page_content}"
        for doc in state["docs"]
    )

    if not os.getenv("OPENAI_API_KEY"):
        return {"answer": context}

    llm = ChatOpenAI(model="gpt-4.1-mini", temperature=0)

    prompt = f"""
    다음 PDF 검색 결과만 참고해서 질문에 답하세요.
    반드시 근거 페이지를 표시하세요.
    근거가 없으면 '확인 불가'라고 답하세요.

    자료:
    {context}

    질문:
    {state['query']}
    """

    response = llm.invoke(prompt)

    return {"answer": response.content}
```

```python
graph = StateGraph(State)

graph.add_node("retrieve", retrieve)
graph.add_node("answer", answer)

graph.add_edge(START, "retrieve")
graph.add_edge("retrieve", "answer")
graph.add_edge("answer", END)

app = graph.compile()

result = app.invoke({
    "query": "AI 에이전트 관련 주요 동향은?",
    "mode": "bm25"
})

print(result["answer"])
```

```python
questions = [
    "AI 에이전트 관련 주요 동향은?",
    "EU AI 법의 투명성 의무는?",
    "오늘 저녁으로 뭐가 좋을까"
]

for question in questions:
    print(f"  ===  [질문: {question}]  ===\n")
    modes = ["bm25", "vector"]

    if vector_retriever is not None:
        modes.append("vector")

    for mode in modes:
        result = app.invoke({
            "query": question,
            "mode": mode
        })

        pages = [
            doc.metadata["page_display"]
            for doc in result["docs"]
        ]

        print("검색 방식:", mode)
        print("검색 페이지:", pages)
        print("답변:", result["answer"][:300])
        print("-" * 40, "\n\n\n")
```
