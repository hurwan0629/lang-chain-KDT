# cursor 생성
- 하나의 데이터베이스 connection에 대해 독립적으로 SQL문을 실행할 수 있는 작업 환경을 제공하는 객체
- 하나의 connection에 동시에 한개의 cursor만 생성 가능, cursor을 통해 SQL문을 실행하면 실행 결과를 튜플단위로 반환

```python
cur = db.cursor()
```

```python
cur.execute("select * from orders")
```

# SQL문의 결과 반환
- fetchall(): 한번에 모든 tuple을 가져옴
- fetchone(): 한번에 하나의 tuple을 가져옴. 다시 fetchone() 메서드를 호출하면 다음 데이터를 가져옴

```python
cur.fetchone()
```

```python
cur.fetchall()
```

```python
cur.fetchone()
```

# 데이터 삽입하기

```python
db = MySQLdb.connect(host='localhost', user='root', password='1234', db='ai')
cur = db.cursor()
```

```python
sql = "insert into member(userid, userpw, name, hp, email, gender, ssn1, ssn2) values (%s, %s, %s, %s, %s, %s, %s, %s)"

data = ('mange', '4352', "망망", "123-1423-1351", 'email', "남", "서울시", "강남구")

cur.execute(sql, data)
db.commit()
cur.execute("select * from member")
cur.fetchall()
```
