# 1. Object Detection
Object Detection(객체 탐지)은 이미지나 영상에서 특정 객체의 존재 여부를 확인하고, 해당 객체의 위치를 바운딩 박스(bounding box)로 표시하는 기술입니다.

이는 컴퓨터 비전에서 중요한 분야로, 이미지 내에서 여러 개의 객체를 동시에 탐지하고 분류할 수 있습니다.

Object Detection은 주로 딥러닝 기반의 CNN(합성곱 신경망) 모델을 활용하며, 대표적인 알고리즘으로는 R-CNN 계열(Faster R-CNN, Mask R-CNN), YOLO(You Only Look Once), SSD(Single Shot MultiBox Detector) 등이 있습니다.

이러한 기술은 자율 주행, 보안 감시, 의료 영상 분석, 증강 현실 등 다양한 분야에서 활용됩니다.

segmentation과 은 방식은 걸리는 시간과 비용이 더 들기때문에 bounding box를 빠르게 처리하고 속도감있게 구하기 위해서 사용하는 방식.

### 1. 객체 탐지의 출력

### 2. 대표적인 딥러닝 기반 계열

1. One-Stage
    - 한번의 네트워크 흐름에서 객체 클래스와 위치를 직접 예측
    - 일반적으로 속도가 빠르며, 실시간 응용에 많이 사용
    - YOLO, SSD, RetinaNet 등
2. Two-stage
    - 먼저 객체 후보 영역을 만들고, 그 후보를 다시 분류하고 박스를 보정함
    - 일반적으로 정밀한 탐지에 강점이 있음.
    - Faster R-CNN, Mask R-CNN 등

> 실무에서는 자율주행, CCTV 분석, 제조 불량 검출, 의료영상 분석 등에서 사용. 현대 모델에서는 "One-Stage"는 조금 부정확하고 "Two-Stage"는 정확하다와 같은 단정을 하면 안됨. 데이터셋, 모델 크기, 학습 방식에 따라 성능 관계가 달라질 수 있음.

# 4. YOLO
YOLO(You Only Look Once)는 이미지를 격자 단위로 나누어 각 격자에서 객체의 위치(바운딩 박스 좌표)와 클래스를 동시에 예측하는 1단계 객체 탐지(One-Stage Detection) 알고리즘입니다.

한 번의 신경망 추론으로 전체 이미지의 탐지를 수행하기 때문에 매우 빠르고, 실시간 객체 탐지에 적합하다는 장점이 있습니다.

YOLO 계열은 v1에서 시작해 v3까지는 Joseph Redmon이 개발했으며, 이후 v4·v7은 커뮤니티 연구자들, v5·v8·v11은 Ultralytics가 주도적으로 발전시켜 현재까지 이어지고 있습니다.

이러한 발전 과정을 거치면서 YOLO는 속도와 정확도의 균형을 잡은 대표적인 객체 탐지 모델로 자리 잡아 다양한 산업 현장에서 활용되고 있습니다.

Yolo1의 경우에는 하나의 이미질ㄹ 여러가지 경계로 나누어 다시 누는 방식을 사용하였습니다.

이후에 점점 발전하면서 빠르고, 실시간 객체탐지가 가능하게 만들어가고 있습니다.

실제로 5, 8, 11 버전이 빠르고, 정확도 높은 대표적인 객체탐지 모델로 자리잡았습니다. (다른 버전은 정확도가 아쉬운 면이 존재하며, 5, 8, 11은 상업용으로 라이선스비를 내야하는 요소입니다.)

### 1. YOLO 모델 크기별 분류
- n (nano): 초경량 모델로 모바일 및 임베디드 장치에서 사용하기 적합합니다.
- s (small): 속도가 빠르면서도 정확도가 준수하여 실시간 추론이 필요한 경우 사용합니다.
- m (medium): 균형 잡힌 모델로 중간 정도의 성능과 속도를 제공합니다.
- l (large): 높은 정확도를 요구하는 애플리케이션에서 사용합니다.
- x (extra-large): 최대 성능을 발휘하지만, 속도가 상대적으로 느립니다.

# 2. YOLO 모델 유형별 분류
- YOLO Detect: 가장 일반적인 객체 탐지 모델로, 특정 객체의 위치와 크기를 바운딩 박스로 반환합니다.

# **4. YOLO annotation 기본 구조**

```
<class_id>, <x_center>, <y_center>, <width>, <height>
   6        0.670220    0.862387    0.573920  0.272823
```
- class_id: 0부터 시작하는 클래스 번호
- x_center, y_center: 박스의 중심점
- width, height: 박습의 너비와 높이
- 좌표는 모두 이미지 크기로 나눈 0~1 사이 값

> VOC XML의 (xmin, ymin, xmax, ymax)를 YOLO 형식으로 바꿀 때에는 중심점과 크기를 계산한 뒤, 이미지 너비와 높이로 정규화 해야함.

````python
VOC_CLASSES = [
    'aeroplane', 'bicycle', 'bird', 'boat', 'bottle',
    'bus', 'car', 'cat', 'chair', 'cow',
    'diningtable', 'dog', 'horse', 'motorbike', 'person',
    'pottedplant', 'sheep', 'sofa', 'train', 'tvmonitor'
]
````

````python
CLASS_TO_ID = {name: idx for idx, name in enumerate(VOC_CLASSES)}
CLASS_TO_ID
````

````python
import xml.etree.ElementTree as ET
````

````python
def voc_box_to_yolo(xmin, ymin, xmax, ymax, image_w, image_h):
    x_center = round((xmin + xmax) / 2 / image_w, 6)
    y_center = round((ymin + ymax) / 2 / image_h, 6)

    x_center *= -1 if x_center < 0 else 1
    y_center *= -1 if y_center < 0 else 1

    w = round((xmax - xmin) / image_w, 6)
    h = round((ymax - ymin) / image_h, 6)

    return x_center, y_center, w, h
````

````python
def convert_cov_xml(xml_path, label_path):
    # xml 파일에 접근할 수 있는 루트
    root = ET.parse(xml_path).getroot()
    size = root.find("size")

    image_w = float(size.findtext("width"))
    image_h = float(size.findtext("height"))

    lines = []

    for obj in root.findall("object"):
        class_name = obj.findtext("name")

        # class_name가 등록되지 않은 ID라면 넘어가주기
        if class_name not in CLASS_TO_ID:
            continue

        bnd = obj.find('bndbox')
        xmin = float(bnd.findtext("xmin"))
        ymin = float(bnd.findtext("ymin"))
        xmax = float(bnd.findtext("xmax"))
        ymax = float(bnd.findtext("ymax"))

        # width, height, x_center, y_center가 YOLO 어노테이션을 찾기 위해 필요
        xmin = max(0.0, min(xmin, image_w))
        xmax = max(0.0, min(xmax, image_w))
        ymin = max(0.0, min(ymin, image_h))
        ymax = max(0.0, min(ymax, image_h))

        if xmax <= xmin or ymax <= ymin:
            continue

        x, y, w, h = voc_box_to_yolo(xmin, ymin, xmax, ymax, image_w, image_h)

        class_id = CLASS_TO_ID[class_name]
        lines.append(f"{class_id} {x:.6f} {y:.6f} {w:.6f} {h:.6f}")

    label_path.parent.mkdir(parents=True, exist_ok=True)
    label_path.write_text("\n".join(lines), encoding="utf-8")
````

### 1. Epoch
전체 학습 데이터를 한 번 모두 학습한 것

### 2. GPU_mem
- 현재 학습 과정에서 사용 중인 GPU 메모리(VRAM)와 관련된 값

### 3. box_loss
- 정답 박스가 있는데 모델이 다른 위치에 박스를 그렸다면 손실이 발생함. (모델이 객체의 위치를 얼마나 잘 찾고 있는지에 대한 손실값.)

### 4. cls_loss
객체 종류를 얼마나 잘 맞히는지와 관련된 손실

### 5. dfl_loss
- dfl(Distribution focal loss)
- Bounding Box의 경계 위치를 보다 정밀하게 예측하기 위한 Loss

> 박스 전체 면적보다는 경계를 얼마나 정확하게 보여주는지에 대한 이야기.
- polygon, linestring과의 차이 정도라고 확인할 수 있음.

### 6. Instances
- 이미지 개수가 아니라 객체의 개수를 나타냄.
- 한 이미지 안에도 여러 객체가 있을 수 있음.

### 7. Size
모델 학습에 사용하는 입력 이미지의 크기

### 8. Epoch가 끝나면 나오는 Validation 결과

```
class: 전체 클래스를 종합해서 평가한 결과
Images: 검증에 사용된 이미지 수
Instances: 검증 이미지 안에 존재하는 정답에 대한 객체 개수
Precision: YOLO가 있다고 한 것 중에서 진짜가 얼마나 있는지
Recall: 실제로 존재하는 객체를 얼마나 놓치지 않고 찾았는가? 놓친 객체가 적을 수록 높아짐
mAP50: IoU 0.5를 기준으로 평가. 정답 박스와 예측 박스가 어느정도 겹치는지 계산
mAP50-95: IoU 0.50 ~ 0.95 까지 모두 순서대로 AP를 계산한 뒤에 평균을 내는 값
```

# **7. 객체 탐지 성능 평가 지표**

ai는 task마다 (결과물을 낼 때 마다) 평가지표가 굉장히 많고, 프로젝트가 잘 진행되고 있는지에 대해서 이야기하고 있는데, 객체탐지도 객체탐지만의 평가지표가 존재함.

당연히 객체탐지는 단순한 accuracy만으로 성능을 평가하기 어려움.

영상이 있는데 영상에 사람도 있고, 자동차도 있는 것에 대해서 사람도 잘 찾고, 자동차도 잘 찾는 것에 대해서 이야기를 하는 부분에 있음.

accuracy로는 표현을 할 수 없음.

bounding box로 추론을 할 때 bounding box가 객체에 정확하게 쳐지지 않을 수 있음.

애매하게 걸쳐진 정답이 있을 수 있기 때문에 정확도지표를 사용하기 어려움

분류 문제에서는 보통 클래스가 맞았는지 확인하면 되지만, 객체 탐지에서는 두가지를 동시에 확인해얗,ㅁ.

1. 예측 클래스가 정답 클래스와 같은지를 비교 - 이거는 정확도를 이용해서 답을 구할 수 있음
2. 예측한 Bounding Box가 실제 객체 위치와 충분히 겹치는지 확인해야함. 완벽하게 감싸는게 맞는지, 클래스가 큰게 맞는 것인지 등을 확인할 필요가 있음.

> Confidence Score가 높을수록 모델이 해당 탐지를 더 강하게 믿고 있다는 의미

### 1. IoU (Intersection over Union)
IoU는 객체 탐지에서 예측 Bounding Box와 정답 Bounding Box의 위치가 얼마나 비슷한지 측정하는 지표
- Intersection: 겹치는 부분 (교집합): 예측박스와 정답박스가 실제로 겹치는 영역
- 이미지에서 박스를 치는데 얼마나 정확하게 표시를 하고 있는지를 보여주는 것을 말함.


어노테이션에 이미 라벨링이 되어있기 때문에 괜찮고, 나중에 겹쳐지는 영역에서 합쳐지게 됩니다.
- Union(합집합): 두 박스를 합쳤을 때 차지하는 전체 영역

> 객체 탐지에서 클래스가 맞아도 Bounding Box가 객체와 거의 겹치지 않는다면 올바른 탐지라고 보기 어려움. 따라서 평가할 때 일정 IoU이상을 만족한 예측만 TP로 인정함.

IoU = 교집합 면적 / 합집합 면적으로 만들어지게 됩니다.

```
IoU = 1# 예측 박스와 정답 박스가 완전히 일치

IoU가 1에 가까우면 # 위치를 매우 정확하게 예측한다는 의미를 가집니다.

IoU가 작으면 # 두 박스가 조금만 겹치거나 위치가 많이 어긋난다는 것을 의미합니다.

```

> 평가할 때 일정 Iou 이상을 만족한 예측만 TP라고 하는데

> AP50의 50은 대표적으로 IoU 0.50을 기준으로 TP/FP를 판단하여 AP를 계산한다는 의미를 가짐.

여기에서 50이라는 의미 자체가 IoU를 0.50으로 맞추었다를 의미함.

AP란? AveratePositive를 말해서 평균 맞는 양을 말하는 것으로 보입니다.

### 2. Confidence Score
- 모델이 이 예측에 대해 얼마나 믿고 있는가를 나타내는 점수
- 예) person 0.92 > 이 곳에 사람이 있다고 판단했고, 이 예측에 대한 신뢰도가 높다
- YOLO는 이미지에서 딱 하나의 박스만 만드는 것이 아님. 객체일 가능성이 있는 많은 후보 박스를 예측

사람이 있을 때 바운딩 박스를 엄청나게 많이 박스를 만들 때, 아래와 같은 여러개의 박스를 칠 수 있음
```
후보1 > person 0.94
후보2 > person 0.85
후보3 > person 0.64
후보4 > dog 0.08
```

```
- confidence threshold (신뢰도 임계값)을 사용해서 어느정도 수준 이하의 정확도의 예측은 버리게 됨. 이를 통해 믿을만한 예측만을 걸러내게 됨.
```

confidence score을 내는 방식은 모델들의 버전마다 다름

### AP(Average Precision)
- 한 클래스에 대해 모델의 Precision과 Recall 성능을 하나의 값으로 요약한 지표
- 객체 탐지에서는 각 예측마다 confidence score가 있음. 평가기는 높은 confidence 예측부터 낮은 confidence 예측까지 순서대로 보며 Precision과 Recall의 변화를 계산. 이를 이용해서 PR 커브를 만들 수 있음.
    - 가로축: Recall
    - 세로축: Precision
- AP 계산
    1. 해당 클래스의 모든 예측 결과를 모음
    2. confidence가 높은 순서로 정렬
    3. 각 예측을 정답 박스와 비교
    4. 클래스와 IoU 조건을 이용해서 TP 또는 FP를 결정
    5. 예측을 하나씩 포함시키면서 Precision과 Recall을 계산
    6. Precision-Recall Curve
    7. 전체값을 요약해서 AP를 계산
- AP50: IoU 기준을 0.50으로 두었을 때 계산한 AP
- AP75: IoU 기준을 0.75으로 두었을 때 계산한 AP

> IoU 기준이 높아질수록 Bounding Box의 위치가 더 정확해야 TP로 인정됨. 따라서 AP75가 AP50보다 더 엄격한 평가

### mAP(mean Average Precision)

객체 탐지 데이터셋에는 여러 클래스가 있으므로, 클래스별 AP를 구한 뒤 평균을 내어 전체의 모델 성능을 표현한 값

- mAP: IoU=0.50 기준에서 각 클래스의 AP를 구한 뒤 클래스 전체 평균을 낸 값. Bounding Box가 정답과 어느정도만 겹쳐도 TP로 인정되기 때문에 상대적으로 점수가 높게 나올 수 있음.
- mAP50-95: IoU 기준을 0.50, 0.55, 0.60, ... 0.90, 0.95 각 IoU 기준에서 AP를 계산하고, 클래스와 IoU 기준 전체에 대해 평균을 냄

> mAP50의 경우에는 그냥 객체를 찾기는 하는지
> map50-95는 단순히 객체를 대략 찾는 것뿐 아니라 Bounding Box의 위치를 얼마나 정밀하게 맞추는지까지 더 엄격하게 반영

- 평가
    - mAP50은 높은데 mAP50-95가 상대적으로 많이 낮음 > 객체 자체는 잘 찾지만 Bounding Box위치 정밀도가 부족할 가능성
    - 둘 다 낮음 > 분류, 위치 예측, 데이터 품질, 학습량 등 전체적인 점검 필요

> 전체 AP를 보는 것이 아닌 특정 클래스만 낮게 볼 수 있기 때문에 AP 또한 따로 볼 필요가 존재한다.

- Confidence threshold는 박스를 치기 전에 모델이 그것이 객체라고 판단을 강하게 확신해야만 찍게 하는 임계점 기준임.
- IoU Threshold는 박스가 얼마나 정밀하게 예측을 하였는지를 보여주는 내용에 대해서 나타내는 값이다.

## NMS
YOLO에 predict(..., iou=...) 또는 검증 설정에서 등장하는 iou는 겹쳐있는 예측 박스를 정리하는 NMS용 threshold

- 평가 IoU: 예측 박스와 정답 박스를 비교
- NMS IoU: 여러 예측 박스끼리 비교하면서 중복 탐지를 제거

박스끼리 겹쳐잇는 것들을 listup해서 많이 겹쳐있는 것을 하나만 선택하는 것을 NMS 면적 값에서 하나만 선택하는 것을 말함.

### FPS와 Latency
- FPS(Frames Per Second): 1초에 처리 가능한 프레임 수
- Latency: 한 입력이 결과로 나오기까지 걸리는 시간
