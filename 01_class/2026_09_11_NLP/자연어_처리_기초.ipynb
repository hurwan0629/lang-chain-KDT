# **3. Tokenization**
- 텍스트를 모델이 처리할 수 있는 작은 단위인 Token으로 나누는 과정
- [토큰 계산](https://platform.openai.com/tokenizer)

### 토큰
- 토큰은 모델이 처리하는 텍스트의 작은 단위
- 하나의 토큰은 단어, 부분적인 단어, 또는 공백이나 구두점일 수 있음
- 토큰은 반드시 단어 하나를 의미하지는 않음

### Vocabulary
- Vocabulary는 Tokenizer가 사용하는 Token들의 집합을 의미함
- 각 토큰에는 고유한 정수 ID가 할당됨
- 정수 ID 자체는 언어적인 의미가 전혀 없음

### 허길 페이스
허깅 페이스(Huffing Face)는 자연어 처리 뿐 아니라 이미지, 음성, 영상, 멀티모달 모델을 공유하고 사용하는 데 필요한 도구와 서비스를 제공하는 AI 생태계이자 회사 이름

````python
if torch.cuda.is_available():
    DEVICE = torch.device('cuda')
elif torch.backends.mps.is_available():
    DEVICE = torch.device('mps')
elif torch.xpu.is_available():
    DEVICE = torch.device('xpu')
else:
    DEVICE = torch.device('cpu')

print(DEVICE)
````

````python
text = "나는 자연어를 공부한다"
tokens = text.split(" ")

# tokens
vocab = {"<PAD>": 0, "<UNK>": 1, "나는:": 2, "자연어를": 3, "공부한다": 5}
token_ids = [vocab.get(token, vocab["<UNK>"]) for token in tokens]
token_ids

input_ids = torch.tensor(token_ids, dtype=torch.long)
````

````python
print("원문: ", text)
print("Token: ", tokens)
print("Token ID: ", token_ids)
print("PyTorch Tensor: ", input_ids)
print("dtype: ", input_ids.dtype)
````

### Token ID
Tokenization이 끝나면 각각의 Token을 Vocabulary에 정의된 정수 ID로 변환
```
"오늘 날씨가 좋다" > Tokenization > [Token1, Token2, Token3. ...] > Token ID > [101, 205, 421, 23, ...]
```

m### Padding

- 문장은 각각 길이가 다름
- **Tensor**의 크기를 맞추기 위해 짧은 `Sequence`에 특별한 Padding Token을 추가하게 됨.

```python
문장 1 [15, 23, 51]
문장 2 [83, 12, 94, 21, 55]

▼ ▼ Padding 후 ▼ ▼
문장 1 [15, 23, 51, 0 , 0 ]
문장 2 [83, 12, 94, 21, 55]
```

> 일반적으로 0을 `<PAD>`의 ID로 사용
>

````python
from torch.nn.utils.rnn import pad_sequence

# 문장 텐서
sentence1 = torch.tensor([15, 23, 51], dtype=torch.long)
sentence2 = torch.tensor([83, 12, 94, 21, 55], dtype=torch.long)
padded = pad_sequence(
    [sentence1, sentence2],
    padding_value=0,
    batch_first=True # 문장을 batch 형식대로 다시 만들어줌 (이렇게 하지 않으면 token 배열이 열에 나열되게 됨)b
)
print(padded, "\n", padded.shape)
````

# **4. 벡터화**
벡터화는 텍스트 데이터를 숫자 형태로 변환하여 머신러닝 또는 딥러닝 모델에서 처리할 수 있도록 만드는 과정입니다. 이를 위해 단어의 빈도를 기반으로 한 Bag-of-Words(BOW), TF-IDF와 같은 방법부터, 단어 간의 의미적 관계를 학습하는 Word Embedding(단어 임베딩), 문장의 문맥과 구조를 반영하는 Transformer 기반 임베딩(BERT, GPT) 등 다양한 기법이 사용됩니다. 벡터화는 단어, 문장, 문서와 같은 텍스트 데이터의 특징을 수치적으로 표현하며, 단순히 단어의 빈도를 반영하거나 의미적 유사성을 학습하여 벡터 공간에서 단어 간 관계를 나타냅니다. 데이터의 특성과 분석 목적에 따라 적절한 벡터화 기법을 선택하는 것이 중요합니다.

````python
import torch.nn.functional as F

token_id = torch.tensor([2, 2, 2, 1, 3, 4], dtype=torch.long)
vocab_size = 5
one_hot = F.one_hot(token_id, num_classes=vocab_size)

token_id, one_hot
````

# 불용어
사용하지 않을 단어는 빼는 것을 불용어라고 부름m

````python
train_data = train_data[:20000]
````

````python
# 샘플 불용어 리스트
stopwords = ['도', '는', '다', '의', '가', '이', '은', '한', '에', '하', '고', '을', '를', '인', '듯', '과', '와', '네', '들', '듯', '지', '임', '게']
````

````python
# train_data["document"] = train_data["document"].str.replace('[^ㄱ-ㅎㅏ-ㅣ가-힣]', '', regex=True)
train_data['document'] = train_data['document'].str.replace('[^ㄱ-ㅎㅏ-ㅣ가-힣 ]', '', regex=True)

train_data[:5]
````

````python
train_data["document"].dtype
````

````python
macab = MeCab()
tokenized_data = []

for sentence in train_data["document"]:
    tempX = mecab.morphs(sentence)
    tempX = [
        word
        for word in tempX
        if not word in stopwords
    ]

    tokenized_data.append(tempX)
````

````python
from gensim.models import Word2Vec
````

````python
# sentences: 학습에 사용할 문장 데이터
# vector_size: 벡터 표현
# window: 주변 단어를 몇 개 볼 것인지
# min_count: 몇 번 이상 등장한 단어만 학습에 포함
# workers: CPU 코어 사용 개수
# sg: CBOW(0), Skip-gram(1)
# negative: Negative Sampling에서 사용할 음성 샘플 개수
model = Word2Vec(
    sentences=tokenized_data,
    vector_size=100,
    window=5,
    min_count=5,
    workers=2,
        sg=1,
    negative=5
)
````

````python
model.wv.vectors.shape
````

````python
# Word2Vec이 학습한 단어 중에서 의미적으로 비슷하게 사용된 단어를 찾아줌
# wv: 학습된 단어 벡터들을 관리
# 결과는 코사인 유사도 > 1에 가까울수록 두 단어의 벡터 방향이 비슷함
model.wv.most_similar("블록버스터", topn=5)
````
