# 2. 자연어 처리
자연어 처리는(Natural Language Processing, NLP) 컴퓨터가 인간의 언어를 이해하고 처리하며 생성할 수 있도록 돕는 인공지능 기술 분야입니다.

# **3. Corpus와 Document**

- Corpus(말뭉치)는 자연어 처리에서 분석하거나 학습하기 위해 수집한 전체 텍스트 데이터의 집합
- Document라고 해서 반드시 긴 문서를 의미하는 것은 아님
- 예) 뉴스기사 하나, 리뷰 하나, 댓글 하나, 문장 하나 등

# **3. Tokenization**
- 텍스트를 모델이 처리할 수 있는 작은 단위인 Token으로 나누는 과정
- [토큰 계산](https://platform.openai.com/tokenizer)

### 토큰
- 토큰은 모델이 처리하는 텍스트의 작은 단위
- 하나의 토큰은 단어, 부분적인 단어, 또는 공백이나 구두점일 수 있음
- 토큰은 반드시 단어 하나를 의미하지는 않음

### Vocabulary
- Vocabulary는 Tokenizer가 사용하는 Token들의 집합을 의미함
- 각 토큰에는 고유한 정수 ID가 할당됨
- 정수 ID 자체는 언어적인 의미가 전혀 없음

### 허길 페이스
허깅 페이스(Huffing Face)는 자연어 처리 뿐 아니라 이미지, 음성, 영상, 멀티모달 모델을 공유하고 사용하는 데 필요한 도구와 서비스를 제공하는 AI 생태계이자 회사 이름

````python
import torch

print(torch.__version__)
````

````python
if torch.cuda.is_available():
    DEVICE = torch.device('cuda')
elif torch.backends.mps.is_available():
    DEVICE = torch.device('mps')
elif torch.xpu.is_available():
    DEVICE = torch.device('xpu')
else:
    DEVICE = torch.device('cpu')

print(DEVICE)
````

````python
text = "나는 자연어를 공부한다"
tokens = text.split(" ")

# tokens
vocab = {"<PAD>": 0, "<UNK>": 1, "나는:": 2, "자연어를": 3, "공부한다": 5}
token_ids = [vocab.get(token, vocab["<UNK>"]) for token in tokens]
token_ids

input_ids = torch.tensor(token_ids, dtype=torch.long)
````

````python
print("원문: ", text)
print("Token: ", tokens)
print("Token ID: ", token_ids)
print("PyTorch Tensor: ", input_ids)
print("dtype: ", input_ids.dtype)
````

````python
# 허깅페이스에서 제공하는 모듈
from transformers import AutoTokenizer
````

````python
# 허깅페이스의 허브의 토큰 설정과 vocabulary를 가져오는 허브
# bert는 무언가가 있는 것이다.
tokenizer = AutoTokenizer.from_pretrained("klue/bert-base")
````

딥러닝, CNN 류의 모든 연산이 뭔지 이미지를 파악하기 위해서, 데이터를 학습시키기 위해서 행렬 연산을 함.

행렬 연산은 사이즈가 맞아야함. 자연어도 행렬 연산을 해야하는데 사이즈가 맞아야만 함

사이즈를 넘기게 되면 다음번으로 넘기게 됨. (일정 량 이 상의 크기를 token을 하지 않고, 크기가 작으면 <PAD>를 넣어주는 방식을 사용함)

vocabulary에 스페셜 토큰이 존재함. (PAD: 크기 키우기용, UNK: unknown)

````python
tokenizer
````

````python

tokens = tokenizer.tokenize(text)
print("Token:", tokens)
token_ids = tokenizer.convert_tokens_to_ids(tokens)
print("token ID:", token_ids)
````

### OOV
- OOV(Out-Of-Vocabulary)는 Voca에 존재하지 않는 Token이 입력되는 문제
- 과거의 단어 단위 Voca에서는 처음보는 단어를 \<UNK\>와 같은 Unknown Token으로 처리하는 경우 가 많았음
- 현재 NLP에서는 Subword Tokenization을 이용하여 이 문제를 크게 줄임

````python
vocab = {"<PAD>": 0, "<UNK>": 1, "나는:": 2, "파이토치를": 3, "공부한다": 5}
tokens = {"나는", "트랜스포머를", "공부한다"}
````

### Token ID
Tokenization이 끝나면 각각의 Token을 Vocabulary에 정의된 정수 ID로 변환
```
"오늘 날씨가 좋다" > Tokenization > [Token1, Token2, Token3. ...] > Token ID > [101, 205, 421, 23, ...]
```
