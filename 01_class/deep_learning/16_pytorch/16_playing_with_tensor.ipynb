```python
from torch.utils.data import Dataset

class MyDataset(Dataset):

    def __init__(self, data, label):
        self.data = data
        self.label = label

    def __len__(self):
        return len(self.data)

    def __getitem__(self, index):
        # print(f"MyDataset:getitem:index-{index}")
        return self.data[index], self.label[index]

    # def __repr__(self):
    #     rs = torch.tensor([idx for idx in self])
    #     return sel
```

```python
from torchvision.datasets import ImageFolder
from pathlib import Path

test_dir = Path() / "data" / "shape" / "test"
train_dir = Path() / "data" / "shape" / "train"

test_dir.exists(), train_dir.exists()
```

```python
test_dataset = ImageFolder(test_dir)
train_dataset = ImageFolder(train_dir)

type(test_dataset), type(test_dataset[0]), type(train_dir), type(train_dataset[0]), type(train_dataset[0][0])
```

```python
train_dataset[0][0]
```

```python
from torchvision import transforms

circle_tensor = transforms.ToTensor()(train_dataset[0][0])
values, indices = torch.sort(circle_tensor.unique())

values
```

```python
def test_transform(trans=None):
    if not trans:
        print("transform required")
        return

    image_tensor =  trans(train_dataset[0][0])

    image_tensor = transforms.Grayscale()(image_tensor)

    print(type(image_tensor))

    return transforms.ToTensor()(image_tensor)
```

```python
k = test_transform(transforms.RandomInvert(p=1.0))
k.shape
k.squeeze(0).shape
```

```python
import matplotlib.pyplot as plt

fig, ax = plt.subplots(1, 1, figsize=(4, 4))

ax.imshow(k.squeeze(0).tolist(), vmin=0, vmax=1, cmap="gray")

plt.show()
```
