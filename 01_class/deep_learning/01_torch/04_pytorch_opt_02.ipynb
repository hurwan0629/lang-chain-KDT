```python
import torch
```

```python
device="cuda"

x = torch.tensor(
    [1., 2., 3., 4.],
    device=device
)

y = torch.tensor(
    [3., 5., 7., 9.],
    device=device
)

w = torch.tensor(
    0.0,
    requires_grad=True,
    device=device
)

b = torch.tensor(
    0.0,
    requires_grad=True,
    device=device
)
```

```python
loss_fn = torch.nn.MSELoss()

# 최적화 객체
optimizer = torch.optim.SGD(
    [w, b],
    lr=0.01
)
```

```python
for epoch in range(100):

    # 1. 예측
    y_pred = w * x + b

    # 2. Loss 계산
    loss = loss_fn(y_pred, y)

    # 3. 이전 gradient 제거
    optimizer.zero_grad()

    # 4. gradient 계산
    loss.backward()

    # 중간 관찰
    if epoch < 3:
        print(
            f"""
epoch={epoch}
loss={loss.item():.4f}
w={w.item()}
b={b.item()}
w.grad={w.grad.item():.4f}
b.grad={b.grad.item():.4f}
            """
        )

    # 5. 파라미터 수정
    optimizer.step()
print("w =", w.item())
print("b =", b.item())
```
