```python
import torch
print(torch.__version__)
print(torch.cuda.is_available())
```

```python
x = torch.tensor(3.0)
y = torch.tensor(7.0)

w = torch.tensor(1.0, requires_grad=True)
b = torch.tensor(0.0, requires_grad=True)

y_pred = w * x + b
loss = (y_pred - y) ** 2
# (wx**2 + 2wx(b-y) + (b**2 - 2by + y**2))
# dloss/dw = 2*x**2 * w + 2x(b-y) -> 18 - 42 = -24
# dloss/db = 2b + 2wx - 2y = 6 - 14 = -8

print("y_pred:", y_pred)
print("loss:", loss)
```

```python
loss.backward()
print("x.grad:", x.grad)
print("w.grad:", w.grad)
print("b.grad:", b.grad)
```

```python
w_new = w - w.grad * 0.1
w_new
```

```python
b_new = b - b.grad * 0.1
b_new
```

```python
y_pred_new = w_new * x + b_new
loss = (y_pred_new - y) ** 2
```
