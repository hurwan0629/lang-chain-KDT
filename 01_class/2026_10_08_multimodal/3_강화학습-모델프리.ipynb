# **1. TD Learning**


- 어떤 길이 좋은지 판단하려면 원래는 끝까지 가서 받은 보상을 알아야 함
- 미로가 길거나 게임이 오래 지속되면 기다리는 동안 배울 수 없음
- TD는 한 칸 움직인 직후 예상치를 고침

```
1. A 위치의 미래 점수를 지금 3점으로 예상
2. A에서 B로 이동(즉시 1점을 받음)
3. B에서 앞으로 얻을 점수는 5점이라고 예상
4. 미래 점수를 90%만 반영한다면 새로운 목표는 1 + 0.9*5 = 5.5점
5. 현재 예상 3점과 새 목표 5.5점의 차이는 +2.5점. 한 번의 경험만 믿고 3점을 곧바로 5.5점으로 바꾸면 불안정할 수 있으므로 차이 10%인 0.25점만 반영해 3.25점으로 수정
```


$$V(s) \leftarrow V(s) + \alpha[r + \gamma V(s’) - V(s)]$$

$$
\delta = r + \gamma V(s') - V(s)
$$

- V(s): 이동 전 상태 A에 대해 원래 예상했던 점수 예) 3
- r: 실제 이동 직후 받은 보상 예) 1
- V(s'): 이동 후 상태 B에서 앞으로 받을 것으로 예상하는 점수 예) 5
- $\gamma$: 미래 점수를 현재에 얼마나 반영할지 정하는 할인율 예) 0.9
- $\delta$: 예상과 새 목표의 차이 예) TD 오차 > 2.5

>5.5점은 미래의 실제 정답이 아니라 현재의 추정값을 사용해 만든 임시 학습 목표. 이를 부트스트래핑(bootstrapping)이라고 부름

````python
v_a, v_b = 3.0, 5.0

reward, gamma, alpha = 1.0, 0.9, 0.1

target = reward + gamma * v_b
td_error = target - v_a

new_v_a = v_a + alpha * td_error
print(f'목표={target:.2f}, 오차={td_error:.2f}, 수정된 V(A)={new_v_a:.2f}')
assert abs(new_v_a - 3.25) < 1e-9
````

# 직접 코드 짜기

````python
V = [0.0] * 7
BAD = 0
GOOD = 6
START = 3

REWARDS = {
  (1, 0): 0,
  (5, 6): 1
}

TERMINAL = (BAD, GOOD)

ACTIONS = {
  "left": (0.5, 1),
  "right": (0.5, -1)
}

GAMMA=0.9
ALPHA=0.05
EPISODE_COUNT=300

rng = random.Random(42)

MAX_TRY = 200
````

````python
# 움직임 정의 
def step(state, action):
  next_state = ACTIONS[action][1] + state
  is_done = False
  reward = 0.0
  if next_state in TERMINAL:
    is_done = True

  if (state, next_state) in REWARDS:
    reward = REWARDS[(state, next_state)]

  return next_state, reward, is_done
````

````python
# 몬테카를로
def monte_episode(start=START, actions=ACTIONS, terminal=TERMINAL, limit=MAX_TRY):
  """한번의 에피소드 실행
  actions대로 이동을 한 이후, 다음상태/보상/종료 를 받는다.
  매 이동마다 보상만큼 state를 적용한다.
  반환 값으로는 value 맵을 준다."""
  episode = [] # (state, reward)

  state = start

  # 종료지점 전까지 이동
  while state not in terminal:
    # 이동하기
    action = rng.choices(
      list(actions.keys()),
      weights=list(map(lambda x: x[0], actions.values())),
      k=1
    )[0]

    # 이동하고 결과 받기
    next_state, reward, _ = step(state, action)

    episode.append((state, reward))

    state = next_state

    if len(episode) > limit:
      return None
  # 에피소드 결과 반환
  return episode

def monte_estimate(episode_count: int, v=V, gamma=GAMMA):
  v = V.copy()
  v_history = []
  count = [0]*len(v)

  for _ in range(episode_count):
    episode=monte_episode()
    if episode is None:
      continue

    # 누적 보상
    G = 0
    for state, reward in reversed(episode):
      G = reward + gamma * G
      
      count[state] +=1
      
      v[state] += (G - v[state]) / count[state]

    v_history.append(v.copy())
  
  return v_history
````

````python
monte_v_history = monte_estimate(EPISODE_COUNT)
````

````python
# TD Learning
# 기본 개념 - 이동 -> 업데이트 -> 에피소드 종료
def TD_learning(episode_count: int, v=V, start=START, terminal=TERMINAL, gamma=GAMMA, actions=ACTIONS, alpha=ALPHA, max_iter=MAX_TRY):

  v = v.copy()

  v_history = []

  def td_update(state, next_state, reward):
    target = reward + gamma * v[next_state]
    delta = target - v[state]
    v[state] += alpha * delta
  
  for _ in range(episode_count):
    state = start

    loop = 0
    while state not in terminal:
      loop += 1
      action = rng.choices(
        list(actions.keys()),
        weights=list(map(lambda x: x[0], actions.values())),
        k=1
      )[0]

      next_state, reward, _ = step(state, action)

      td_update(state, next_state, reward)

      state = next_state

      if max_iter < loop:
        break
    v_history.append(v.copy())
  return v_history
````

````python
td_v_history = TD_learning(EPISODE_COUNT)
````

````python
# 실제 가치 구하기 - 벨만 가치 방정식
REAL_V = V.copy()

for _ in range(200):
  for s in range(len(REAL_V)):
    if s in TERMINAL:
      continue

    REAL_V[s] = sum(
      prob * (step(s, key)[1] + GAMMA * REAL_V[s+move])
      for key, (prob, move)
      in ACTIONS.items()
    )

REAL_V
````

````python
import numpy as np

np_td_v_history = np.array(td_v_history)
np_monte_v_history = np.array(monte_v_history)
np_REAL_V = np.array(REAL_V)

np_td_v_history.shape, np_monte_v_history.shape, np_REAL_V.shape
````

````python
td_error = np_td_v_history - np_REAL_V
mt_error = np_monte_v_history - np_REAL_V

import matplotlib.pyplot as plt
````

````python
td_rmse = np.sqrt(np.mean((td_error**2), axis=1))
mt_rmse = np.sqrt(np.mean((mt_error**2), axis=1))
````

# 강사님 코드

````python
def random_walk_episode(rng):
    state=3
    transitions=[]
    while state not in (0, 6):
        action = rng.choice([-1, 1])
        nxt = state + action
        done = nxt in (0, 6)
        reward = 1.0 if nxt == 6 else 0.0
        transitions.append((state, reward, nxt, done))
        state = nxt
    return transitions

def compare_mc_td(episodes=300, alpha=0.05):
    rng = np.random.default_rng(SEED)
    mc = np.zeros(7)
    td = np.zeros(7)
    mc_errors, td_errors = [], []
    truth = np.arange(1, 6) / 6
    
    for _ in range(episodes):
        path = random_walk_episode(rng)

        for s, r, nxt, done in path:
            td[s] += alpha * (r + (0 if done else td[nxt]) - td[s])
        G = 0.0

        for s, r, nxt, done in reversed(path):
            G = r + G
            mc[s] += alpha * (G - mc[s])
        
        mc_errors.append(np.sqrt(np.mean((mc[1:6]-truth)**2)))
        td_errors.append(np.sqrt(np.mean((td[1:6]-truth)**2)))
    return mc, td, mc_errors, td_errors


mc_v, td_v, mc_err, td_err = compare_mc_td()
plt.figure(figsize=(9, 4))
plt.plot(mc_err, label='MC RMSE'); plt.plot(td_err, label='TD RMSE')
plt.xlabel('Episode'); plt.ylabel('RMSE'); plt.legend(); plt.grid(alpha=.3); plt.show()
print('참값:', np.round(np.arange(1,6)/6, 3))
print('MC  :', np.round(mc_v[1:6], 3))
print('TD  :', np.round(td_v[1:6], 3))
````

    - MC: 에피소드가 완전히 종료된 후 실제로 얻은 누적 보상(Return)을 이용하여 상태 가치를 업데이트

- RMSE가 작다는 것은 무엇을 의미하나요?
    - 강화학습에서는 알고리즘이 학습한 상태 가치와 실제 상태 가치가 얼마나 차이 나는지 확인할 때 사용
    - 예) MC가 예측한 가치: 오차 0.02, TD가 예측한 가치: 오차 0.10 > MC 예측이 실제 가치에 더 가까움
    - RMSE가 작을수록 알고리즘이 학습한 상태 가치가 실제 정답에 가깝다는 의미 > RMSE가 0이면 비교한 모든 상태의 예측 가치가 실제 가치와 정확히 일치한다는 뜻
- 실습 결과에서 MC와 TD 중 어느 알고리즘이 더 빠르게 참값에 가까워졌나요? 그 이유는?
    - 일반적으로 Random 실습에서는 TD Learning보다 MC가 더 빠르게 참값에 가까워지는 경향이 있음
    - TD Learning은 매 이동에서 얻은 보상과 다음 상태의 추정 가치를 활용하여 학습 > 에피소드의 최종 결과만을 학습 목표로 사용하는 MC와 달리 이미 학습한 다른 상태의 가치 정보를 현태 상태의 학습에 활용
    - MC는 에피소드가 종료된 후 실제 누적 보상을 이용 > Random 환경에서는 같은 상태에서 출발하더라도 0번 또는 6번에 도착할 수 있기 때문에, 에피소드마다 누적 보상이 0 또는 1로 달라짐 > MC는 학습 목표의 변동성이 상대적으로 클 수 있음

### Grid 환경 구현 및 TD Learning을 이용한 상태 가치 학습

1. 5 * 5 크기의 격자 환경을 구현
- 시작 위치 (0, 0)
- 목표 위치 (4, 4)
- 장애물 위치: (1, 1), (1, 2), (3, 3)
- 행동: 상/하/좌/우
- 목표 도착 보상: +10
- 일반 이동 보상: -0.1
- 에피소드당 최대 이동 횟수: 100
- 로봇은 격자 범위를 벗어나거나 장애물로 이동하려고 하면 현재 위치에 머무름
- 단, 이동하지 못하더라도 이동 횟수는 증가하고 보상 -0.1이 적용
- 목표에 도착하면 에피소드가 종료되며, 최대 이동 횟수에 도달하면 해당 에프소드를 중단

2. 학습 조건
- 에피소드 수: 2000
- 학습률: 0.08
- 할인율: 0.95
- 행동 선택: 상/하/좌/우 무작위 선택
- 상태 가치 초기값: 모두 0

3. 시각화
- 학습된 5 * 5 상태 가치 배열 V
- 각 위치의 상태 가치를 표현한 Heatmap

````python
import random
import numpy as np

# Grid 환경
GRID_SIZE = 5
START = (0, 0)
GOAL = (4, 4)
OBSTACLES = {(1, 1), (1, 2), (3, 3)}

# 행동 정의 (행, 열 변화량)
ACTIONS = {
    "up": (-1, 0),
    "down": (1, 0),
    "left": (0, -1),
    "right": (0, 1)
}

# 보상
GOAL_REWARD = 10.0
STEP_REWARD = -0.1

# 학습 조건
EPISODES = 2000
ALPHA = 0.08
GAMMA = 0.95
MAX_STEPS = 100

# 상태 가치 초기화
V = np.zeros((GRID_SIZE, GRID_SIZE))

# 랜덤 생성기
rng = random.Random(42)
````

````python
# 이동 함수
def step(state, action):
    row, col = state
    dr, dc = ACTIONS[action]

    next_row = row + dr
    next_col = col + dc
    next_state = (next_row, next_col)

    # 격자 밖 또는 장애물이면 이동하지 않음
    if (
        not (0 <= next_row < GRID_SIZE)
        or not (0 <= next_col < GRID_SIZE)
        or next_state in OBSTACLES
    ):
        next_state = state

    # 목표 도착 여부
    done = next_state == GOAL

    # 보상
    reward = GOAL_REWARD if done else STEP_REWARD

    return next_state, reward, done
````

````python
# TD(0) Learning
V_history = {}

for episode in range(1, EPISODES + 1):
    state = START

    for _ in range(MAX_STEPS):
        # 1. 무작위 행동 선택
        action = rng.choice(list(ACTIONS))

        # 2. 환경에서 이동
        next_state, reward, done = step(state, action)

        # 3. TD 목표값 계산
        target = reward + GAMMA * (
            0 if done else V[next_state]
        )

        # 4. TD 오차 계산
        delta = target - V[state]

        # 5. 상태 가치 업데이트
        V[state] += ALPHA * delta

        # 6. 다음 상태로 이동
        state = next_state

        if done:
            break

    # 7. 500 에피소드마다 상태 가치 저장
    if episode % 500 == 0:
        V_history[episode] = V.copy()
````

````python
# 1. Bellman 가치 방정식 (반복적 정책 평가)
V_bellman = np.zeros((GRID_SIZE, GRID_SIZE))

tolerance = 1e-12

while True:
    new_v = V_bellman.copy()

    # 열 이동 -> 행 이동 순서로 전파되기 (4, 4)가 골인지점, (0, 0)부터 순회 시작
    for row in range(GRID_SIZE):
        for col in range(GRID_SIZE):
            state = (row, col)
            # 장애물이나 종료시 패스
            if state in OBSTACLES or state == GOAL:
                continue

            value = 0.0

            for action in ACTIONS:
                # 보상 받아서 상하좌우로 나누어서 가치 갱신해주기
                next_state, reward, done = step(state, action)
                value += 0.25 * (
                    reward + GAMMA * (
                        0 if done else V_bellman[next_state]
                    )
                )

            new_v[state] = value

    # 최대 오차 없을 때까지 반복
    delta = np.max(np.abs(new_v - V_bellman))
    V_bellman = new_v

    if delta < tolerance:
        break


# 2. Linspace 가치맵 (단순 선형 기준)
V_linspace = np.linspace(
    0, GOAL_REWARD, GRID_SIZE * GRID_SIZE
).reshape(GRID_SIZE, GRID_SIZE)
````

````python
class GridWorld:
    ACTIONS = [(-1, 0), (1, 0), (0, -1), (0, 1)]
    def __init__(self, size=5, walls=((1, 1), (1, 2), (3, 3)), max_steps=100):
        self.size = size
        self.walls = set(walls)
        self.goal = (size-1, size-1)
        self.start = (0, 0)
        self.max_steps = max_steps
        assert self.goal not in self.walls and self.start not in self.walls
        self.reset()

    def reset(self):
        self.pos = self.start
        self.steps = 0
        return self.pos

    def step(self, action):
        dr, dc = self.ACTIONS[action]
        r, c = self.pos
        candidate = (r + dr, c + dc)
        if (0 <= candidate[0] < self.size and 0 <= candidate[1] < self.size and candidate not in self.walls):
            self.pos = candidate
        self.steps += 1
        terminated = self.pos == self.goal
        truncated = self.steps >= self.max_steps and not terminated
        reward = 10.0 if terminated else -0.1
        return self.pos, reward, terminated, truncated
````

````python
env = GridWorld()
V = np.zeros((env.size, env.size))
alpha, gamma = 0.08, 0.95
rng = np.random.default_rng(SEED)
for episode in range(2000):
    state = env.reset()
    while True:
        action = int(rng.integers(4))
        nxt, reward, terminated, truncated = env.step(action)
        bootstrap = 0 if terminated else V[nxt]
        V[state] += alpha*(reward + gamma * bootstrap - V[state])
        state=nxt
        if terminated or truncated:
            break
````

    - 현재 상태에서 앞으로 얻을 보상을 아직 정확히 알 수 없기 때문에, 다음 상태에 대해 지금까지 학습한 예상 가치를 이용하여 현재 상태의 가치를 추정하기 위함
- 목표 지점에 가까운 위치와 먼 위치의 상태 가치에는 어떤 차이가 나타나는지. 그 이유는 무엇인지?
    - 일반적으로 목표 지점에 가까운 위치는 먼 위치보다 상태 가지가 높게 나타나는 경향이 있음
- 장애물 주변의 상태 가치는 다른 위치와 어떤 차이가 있는지?
    - 장애물 주변의 상태 가치는 장애물이 없는 위치와 다르게 나타날 수 있으며, 목표로 이동하기 어려운 위치에서는 상대적으로 낮은 상태 가치가 나타나는 경향이 있음
    - 장애물 때문에 이동할 수 있는 경로가 제한되고, 목표까지 우회해야 하거나 제자리에서 시간을 소비할 가능성이 높아지기 때문

# **2. Q-Learning**
최적 행동 가치 Q에 가까워지도록 학습하여, 각 상태에서 Q가 큰 행동을 선택하는 것을 목표로 함

```
로봇이 갈림길 A에 있음
V(A) = 5 : A에 있을 때 앞으로 얼마나 좋은가?를 뜻
현재 왼쪽과 오른쪽 중 뭇엇을 선택할지는 알려주지 못함

Q(A, 왼쪽)=2, Q(A, 오른쪽)=8 : 오른쪽을 고를 수 있음(Q는 상태와 행동을 묶어 평가한 점수)
```

V(s): 이 상태에서 정책을 따라가면 앞으로 받을 예상 보상
Q(s, a): 이 상태에서 특정 행동 a를 먼저 선택하고 이후 정책을 따르면 받을 예상 보상

> 정책에 따라 V와 Q의 값도 달라질 수 있음. Q-Learning은 최적 행동 가치 Q에 가까워지도록 학습하여, 각 상태에서 Q가 큰 행동을 선택하는 것을 목표로 함

### Q-Learning의 계산 과정
1. 지금 (상태, 행동)의 기존 Q점수를 확인
2. 실제 행동을 해 보상을 받음
3. 다음 상태에서 가장 점수가 높은 행동의 Q값을 찾음
4. 지금 받은 보상 + 할인된 다음 상태 최고 점수를 새 목표로 삼음
5. 기존 점수와 목표의 차이를 학습률만큼 반영

````python
q_old=0.0; reward=10.0; alpha=0.1; gamma=0.9
max_next_q=0.0
q_new=q_old+alpha*(reward+gamma*max_next_q-q_old)
print('새 Q(3, 오른쪽):', q_new)
````

````python
def train_line_q(episodes=400, epsilon=0.2, alpha=0.3, gamma=0.9):
    rng = np.random.default_rng(SEED)
    Q = np.zeros((5, 2)) # 행=상태 0~4, 열=행동 0~1
    rewards = []
    for ep in range(episodes):
        state=0
        total=0
        for step in range(30):
            if rng.random() < epsilon:
                action = int(rng.integers(2))
            else:
                best = np.flatnonzero(Q[state] == Q[state].max())
                action = int(rng.choice(best))
            nxt = int(np.clip(state + (-1 if action == 0 else 1), 0, 4))
            done = nxt == 4
            reward = 10 if done else -1
            target = reward + (0 if done else gamma * np.max(Q[nxt]))
            Q[state, action] += alpha * (target - Q[state, action])
            total += reward
            staste = nxt
            if done: break
        rewards.append(total)
    return Q, np.asarray(rewards)
````

````python
Q, line_returns=train_line_q()
print('Q 테이블 (왼쪽, 오른쪽):\n',np.round(Q,2))
print('학습된 정책:', ['←' if a==0 else '→' for a in Q[:4].argmax(axis=1)])
plt.plot(np.convolve(line_returns,np.ones(20)/20,mode='valid'))
plt.title('20-episode moving average return'); plt.xlabel('Episode');plt.show()
````
