# 국민 건강보험공단 사업장 관리 현황
- 이름: `National Health Insurance Service - Status of Managed Workplaces`
- 링크: `https://www.data.go.kr/data/3049051/fileData.do`
- 라이선스: 없음

```python
import pandas as pd
from pathlib import Path

pd.set_option("display.max_columns", None)
pd.set_option("display.max_rows", None)

CSV_PATH = Path() / ".." / ".." / "resource" / "datasets" / "국민건강보험공단_사업장관리_현황_20251231.csv"
NHIS_df = pd.read_csv(CSV_PATH, encoding="cp949")
NHIS_df
```

```python
(NHIS_df.isna().sum() != 0).sum()
```

```python
import matplotlib.pyplot as plt

plt.rcParams["font.family"] = "Malgun Gothic"
plt.rcParams["axes.unicode_minus"] = False

plt.plot(NHIS_df["구분"], NHIS_df["합계"], marker="o")
plt.title("년도별 건강보험공단 관리 사업장 수 추이")
plt.xlabel("연도")
plt.ylabel("사업장 수 (100만)")
plt.xticks(NHIS_df["구분"])
plt.grid(alpha=0.3)
plt.show()
```
